### **Exercise 17. Lowercase First**
Practice Problem: Write a program to arrange string characters such that all lowercase letters come first, followed by all uppercase letters.

Exercise Purpose: This exercise introduces Conditional Filtering and Reassembly. It’s a common pattern in data sorting: you need to group items by a specific property (in this case, “casing”) while preserving their relative order.


Given Input: str1 = "PyNaTive"

Expected Output: yaivePNT

```python
import string
str1 = input("Escriba una palabra: ")
abc_lower = string.ascii_lowercase
abc_upper = string.ascii_uppercase
minusculas = []
mayusculas = []

for char in str1:
    if char in abc_lower:
        minusculas.append(char) #append sirve para agregar algo a las listas
    elif char in abc_upper: 
        mayusculas.append(char)

lista_combinada = minusculas + mayusculas
new_word = "".join(lista_combinada)
print(new_word)
```

Exercise Purpose: This exercise sharpens your ability to handle Multiple Pointer Traversal. It makes you think about coordinating indices moving in opposite directions (forward through one string and backward through another).

Given Input: s1 = "Abc" s2 = "Xyz"

Expected Output: AzbycX

```python
s1 = "Abc" 
s2 = "Xyz"
lista = []
for i in range(len(s1)): #Empieza desde el cero la i 
    lista.append(s1[i])
    lista.append(s2[-(i +1)])

palabra = "".join(lista)
print(palabra)
```

```python
s1 = "Abc"
s2 = "Xyz"

# 1. Obtenemos la longitud (tamaño) de ambas cadenas
s1_length = len(s1)
s2_length = len(s2)

# 2. Determinamos cuál cadena es más larga para que el ciclo 
#    se repita las veces necesarias sin importar si son de diferente tamaño.
length = max(s1_length, s2_length)

result = ""

# 3. Invertimos s2 por completo ([::-1]) al inicio. 
#    Esto evita usar matemáticas complejas con índices negativos dentro del bucle; 
#    ahora el último carácter de s2 pasa a ser el primero.
s2 = s2[::-1]

# 4. Recorremos desde 0 hasta el tamaño de la cadena más larga
for i in range(length):
  # Si la posición 'i' existe en s1, agregamos esa letra al resultado
  if i < s1_length:
    result = result + s1[i]

  # Si la posición 'i' existe en s2 (que ya está invertida), 
  # agregamos la letra correspondiente
  if i < s2_length:
    result = result + s2[i]

print(result)
```

### **Exercise 29. Unique Character Check**
Practice Problem: Write a function to determine if a string has all unique characters. Return True if every character appears only once, otherwise return False.

Exercise Purpose: This introduces the concept of Sets. In Python, a set is an unordered collection that cannot contain duplicate items. Comparing the length of a string to the length of its “set version” is the most efficient way to check for uniqueness in O(n) time.

Given Input: str1 = "python"

str2 = "alphabet"

Expected Output:

"python" unique: True
"alphabet" unique: False

```python
def is_unique(s):
    # A set only keeps unique elements
    char_set = set(s)
    return len(char_set) == len(s)

str1 = "python"
str2 = "alphabet"

print(f"'{str1}' unique: {is_unique(str1)}")
print(f"'{str2}' unique: {is_unique(str2)}")



```

Exercise Purpose: This practice covers String Splitting and Indexing. While .title() exists, manually implementing it helps you understand how to isolate parts of a string, modify them, and rebuild the sentence. This is a common requirement in data transformation.

Given Input: str1 = "hello world from python"

Expected Output: Hello World From Python

```python
str1 = "hello world from python"
str = str1.split()
r = []
longitud = len(str)
for word in str:
    capitalize = word[0].upper() + word[1:]
    r.append(capitalize)

print(" ".join(r))
```

### **Ejercicio 31**

Crear un programa, que tenga una variable con la cadena “Te quiero solo como amigo”, y muestre la siguiente información:

• Imprima los dos primeros caracteres.

• Imprima los tres últimos caracteres.

• Imprima dicha cadena cada dos caracteres. Ej.: Si la cadena fuera “recta” debería imprimir rca

• Dicha cadena en sentido inverso. Ej.: Si la cadena fuera hola mundo! debe imprimir !odnum aloh

• Imprima la cadena en un sentido y en sentido inverso. Ej: Si la cadena es “reflejo” imprime reflejoojelfer.

```python
palabra = "Te quiero solo como amigo"

print(palabra[:2])
print(palabra[-3:])
print(palabra[::2])
print(palabra[::-1])
print(palabra + (palabra[::-1]))
```
