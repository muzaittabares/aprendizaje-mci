## 1. Concepto y Creación

Una cadena (*string*) es una secuencia ordenada e inmutable de caracteres. En Python se representa mediante comillas simples (`' '`), dobles (`" "`) o triples (`''' '''` / `""" """`).

```python

### 2. La Inmutabilidad

Las cadenas en Python son inmutables. Esto significa que, una vez creada una cadena, no puedes modificar directamente uno de sus caracteres.

````python
texto = "python"
#texto[0] = "J" esto daria error

#Para cambiar una cadena debes crear una nueva:
texto = "Python"
nuevo_texto = "J" + texto[1:]
#o
texto = texto.replace("P", "J")

print(nuevo_texto)   # Jython
print(texto)
````

Los métodos de strings tampoco modifican la cadena original; normalmente devuelven una nueva cadena:

````python
texto = "hola"

mayusculas = texto.upper()

print(texto)       # hola
print(mayusculas)  # HOLA
````

### 3. Indexación y Slicing (Rebanados)
Índices: Empiezan a contar desde el 0 (hacia adelante) o desde -1 (la última letra, hacia atrás).

Slicing ([inicio:fin:paso]): El límite de fin nunca se incluye (se detiene uno antes). Si omites partes, Python asume los extremos.

````python
s = "Python"
#La primer indica de donde empieza tomando ese valor, el segundo hasta donde llega pero sin tomar en cuenta ese y el tercero de cuanto en cuanto
print(s[0])      # P (primera letra)
print(s[-1])     # n (última letra)
print(s[0:3])    # Pyt (del 0 al 2)
print(s[::2])    # Pto (toma uno sí, uno no)
print(s[::-1])   # nohtyP (Truco para voltear la cadena)
````

### 4. Operadores Básicos
Concatenación (+) y Repetición (*):
"hola" + " " + "mundo" da "hola mundo" y "ab" * 3 da "ababab".

Pertenencia (in / not in):
"gato" in "el gato negro" da True, y "perro" not in "el gato negro" da True.

Comparaciones (==, <, >): Se comparan alfabéticamente según su valor Unicode. Por ejemplo, "ana" == "Ana" es False (las mayúsculas importan) y "ana" < "beto" es True.

````python
palabra_1 = "hola"
palabra_2 = " mundo"

print(palabra_1 + palabra_2)
print("hola" + " " + "mundo")
print("ab" * 3)

frase = "el gato negro"

print("gato" in frase)
print("perro" not in frase)
print("gato" in "el gato negro")
print("gato" not in "el gato negro")
````

### 5. Iteración (Recorrer un String)
Como las cadenas son objetos iterables, puedes pasearte por cada letra individualmente usando un ciclo for:

````python
for letra in "MCI":
    print(letra) #El ciclo recorre el texto letra por letra de izquierda a derecha, asignando cada carácter a la variable letra para imprimirlo individualmente en cada vuelta.
````

### 7. Métodos Esenciales de los Strings
Se ejecutan colocando un punto después de la variable (variable.metodo()):
#### Tabla de métodos de cadenas en formato Markdown
| Método | Qué hace (Palabras fáciles) | Ejemplo rápido |
| :--- | :--- | :--- |
| **`.find(sub)`** | Busca una letra o palabra y te dice en qué posición (**índice**) empieza. Si no la encuentra, devuelve `-1`. | `"hola".find("l")` $\rightarrow$ `2` |
| **`.count(sub)`** | Cuenta exactamente cuántas veces se repite una letra o fragmento dentro del texto. | `"banana".count("a")` $\rightarrow$ `3` |
| **`.startswith(sub)`** | Revisa si el texto **empieza exactamente** con ese fragmento. Devuelve `True` o `False`. | `"Python".startswith("Py")` $\rightarrow$ `True` |
| **`.endswith(sub)`** | Revisa si el texto **termina** con ese fragmento. Devuelve `True` o `False`. | `"archivo.py".endswith(".py")` $\rightarrow$ `True` |
| **`.upper()`** | Convierte absolutamente todo el texto a **MAYÚSCULAS**. | `"hola".upper()` $\rightarrow$ `"HOLA"` |
| **`.lower()`** | Convierte absolutamente todo el texto a **minúsculas**. | `"HOLA".lower()` $\rightarrow$ `"hola"` |
| **`.strip()`** | Limpia y borra los espacios vacíos sobrantes que estén al **inicio y al final** del texto. | `"  hola  ".strip()` $\rightarrow$ `"hola"` |
| **`.replace(viejo, nuevo)`** | Busca una parte del texto y la reemplaza por otra palabra o letra nueva. | `"gato".replace("g", "p")` $\rightarrow$ `"pato"` |
| **`.split(separador)`** | Troza o divide un texto largo en una **lista** usando un carácter separador (coma, espacio, etc.). | `"a,b,c".split(",")` $\rightarrow$ `['a', 'b', 'c']` |
| **`"pegamento".join(lista)`** | Toma los elementos de una lista y los une en un solo texto usando un "pegamento". | `"-".join(["a", "b", "c"])` $\rightarrow$ `"a-b-c"` |
| **`.isalpha()`** | Valida si el texto está compuesto **únicamente por letras** (sin números ni espacios). | `"Hola".isalpha()` $\rightarrow$ `True` |
| **`.isdigit()`** | Valida si el texto está compuesto **únicamente por números**. | `"12345".isdigit()` $\rightarrow$ `True` |
| **`.capitalize()`** | Pone la **primera letra de todo el texto en mayúscula** y el resto en minúsculas. | `"hola mundo".capitalize()` $\rightarrow$ `"Hola mundo"` |
| **`.title()`** | Pone la **primera letra de cada palabra** en mayúscula (como un título). | `"hola mundo".title()` -> `"Hola Mundo"` |
| **`.swapcase()`** | **Invierte** las letras: las mayúsculas pasan a minúsculas y viceversa. | `"HolaMundo".swapcase()` -> `"hOLAmUNDO"` |
| **`.casefold()`** | Convierte todo a minúsculas de forma agresiva (ideal para comparar textos sin importar tildes o idiomas). | `"HOLA".casefold()` -> `"hola"` |
| **`.isnumeric()`** | Valida si el texto contiene solo números (incluye fracciones o números de otros alfabetos). | `"123".isnumeric()` -> `True` |
| **`.isalnum()`** | Valida si el texto tiene **solo letras y/o números** (sin símbolos ni espacios). | `"Python3".isalnum()` -> `True` |
| **`.isspace()`** | Valida si el texto está compuesto **únicamente por espacios** en blanco. | `"   ".isspace()` -> `True` |
| **`.islower()`** | Valida si **todas las letras** del texto están en minúsculas. | `"hola".islower()` -> `True` |
| **`.isupper()`** | Valida si **todas las letras** del texto están en mayúsculas. | `"HOLA".isupper()` -> `True` |
| **`.istitle()`** | Valida si el texto cumple con formato de título (cada palabra inicia con mayúscula). | `"Hola Mundo".istitle()` -> `True` |
| **`.partition(sep)`** | Parte el texto en tres partes usando un separador: la izquierda, el separador y la derecha. | `"a-b-c".partition("-")` -> `('a', '-', 'b-c')` |
| **`.splitlines()`** | Parte un texto largo en una lista utilizando los **saltos de línea** (`\n`) como guía. | `"hola\nmundo".splitlines()` -> `['hola', 'mundo']` |
| **`.zfill(ancho)`** | Rellena el texto con **ceros a la izquierda** hasta alcanzar el ancho especificado. | `"5".zfill(3)` -> `"005"` |
| **`.ljust(ancho)`** | Justifica el texto a la **izquierda**, rellenando el espacio faltante con espacios o caracteres. | `"hola".ljust(6, "-")` -> `"hola--"` |
| **`.rjust(ancho)`** | Justifica el texto a la **derecha**, rellenando el espacio faltante con espacios o caracteres. | `"hola".rjust(6, "-")` -> `"--hola"` |
| **`.center(ancho)`** | **Centra** el texto, rellenando los espacios sobrantes a ambos lados por igual. | `"hola".center(8, "-")` -> `"--hola--"` |

### 11. Expresiones Regulares (Módulo re)
Las expresiones regulares (Regular Expressions o simplemente Regex) son como un superpoder de búsqueda. En lugar de buscar una palabra exacta (como .find("hola")), te permiten buscar patrones (por ejemplo: "encuentra cualquier secuencia de 5 números" o "busca un correo electrónico válido").

Para usarlas en Python, primero debes escribir import re. Las funciones principales son:

re.search(patron, texto): Revisa el texto y te devuelve el primer fragmento que coincida con el patrón en cualquier parte de la cadena.

re.findall(patron, texto): Busca todas las coincidencias en el texto y te las regresa juntas en una lista.

re.sub(patron, reemplazo, texto): Busca las partes que coincidan con el patrón y las cambia por otra cosa.

````python
import re

# Texto desordenado del cual queremos extraer información importante
texto_sucio = "Contacto: soporte@empresa.com, ventas@tienda.net y código postal 04510."

# Definimos el patrón usando expresiones regulares:
# - r"": Indica que es un raw string (para evitar problemas con barras).
# - \d{5}: Busca exactamente 5 dígitos seguidos (para el código postal).
patron_cp = r"\d{5}"

# Procesamos el texto buscando todas las coincidencias del patrón
resultados = re.findall(patron_cp, texto_sucio)

# Imprimimos el resultado final para ver cómo funciona
print(resultados)
````

### 12. Codificación y Decodificación (encode() y decode() con UTF-8)
¿Cómo lo procesa Python?
Por dentro, Python maneja los textos como cadenas Unicode (str), que son ideales para que los humanos las lean (con tildes, eñes, emojis, etc.). Sin embargo, las computadoras, los discos duros y internet no entienden letras, solo entienden ceros y unos (bytes).

Cuando aplicas .encode("utf-8"), Python traduce cada letra del texto humano a una secuencia exacta de números binarios usando el estándar UTF-8, transformándolo en un objeto de tipo bytes (por eso aparece con una b al inicio).

Cuando aplicas .decode("utf-8"), haces lo opuesto: tomas esos bytes crudos y los reconstruyes en un texto legible. Esto es vital para evitar los molestos símbolos raros o signos de interrogación (``) cuando abres archivos o bases de datos locales que tienen acentos o caracteres especiales.

````python
# Texto original con acentos y eñes (muy común en documentos y facturas en México)
texto_original = "Factura de México: costos, IVA y cédula fiscal"

# 1. ENCODE: Pasamos de texto humano (str) a bytes (utf-8)
# Esto convierte el texto en datos binarios listos para viajar por red o guardarse.
datos_en_bytes = texto_original.encode("utf-8")

print("--- 1. Datos en bytes ---")
print(datos_en_bytes)  # Imprime con la 'b' inicial indicando que son bytes puros

# 2. DECODE: Pasamos de bytes de vuelta a texto humano (str)
# Esto rescata la información para que se vuelva a leer con tildes y eñes perfectamente.
texto_restaurado = datos_en_bytes.decode("utf-8")

# Imprimimos el resultado final para ver cómo funciona
print("\n--- 2. Texto recuperado ---")
print(texto_restaurado)
````

### **SETS**

````python
# 1. EL PODER OCULTO: Eliminar duplicados automáticamente
# Imagina una lista de correos donde alguien se inscribió dos veces por error
correos = ["ana@mail.com", "carlos@mail.com", "ana@mail.com", "sofia@mail.com"]

# Al transformarla en set, los duplicados desaparecen mágicamente
correos_unicos = set(correos)
print("Lista limpia de duplicados:", correos_unicos)


# 2. INTERSECCIÓN: Encontrar elementos en común
# Lenguajes que domina el Programador A y el Programador B
habilidades_a = {"Python", "SQL", "Excel"}
habilidades_b = {"Python", "JavaScript", "HTML"}

# Usamos .intersection() para ver qué saben hacer ambos
en_comun = habilidades_a.intersection(habilidades_b)
print("Habilidades que ambos comparten:", en_comun)  # Resultado: {'Python'}
````

````python
# ==========================================
# GUÍA PRÁCTICA DE SETS (CONJUNTOS) EN PYTHON
# ==========================================

# 1. ¿QUÉ ES UN SET Y CÓMO SE CREA?
# Un set es una colección desordenada de elementos ÚNICOS (sin duplicados).
# Si pasas una palabra o lista con elementos repetidos, Python los elimina automáticamente.
palabra = "programacion"
letras_unicas = set(palabra)
print("1. Set desde string (elimina duplicados y desordena):", letras_unicas)


# 2. EL MÉTODO .issubset() (El que usamos para validar palabras)
# Sirve para responder a la pregunta: ¿TODO el contenido del primer set 
# se encuentra dentro del segundo set? (El orden no importa).
s1 = set("yn")         # Letras obligatorias a buscar: {'y', 'n'}
s2 = set("PyNative")   # Bolsa de letras general: {'P', 'y', 'N', 'a', 't', 'i', 'v', 'e'}

# Comprobamos si 'yn' es un subconjunto de 'PyNative'
es_valido = s1.issubset(s2)
print("2. ¿Es subconjunto (.issubset):", es_valido)  # Devuelve True porque 'y' y 'n' están ahí


# 3. .union() (UNIR DOS CONJUNTOS)
# Junta dos sets en uno solo, fusionando sus elementos y eliminando los repetidos.
lenguajes_a = {"Python", "SQL"}
lenguajes_b = {"SQL", "JavaScript", "HTML"}

# Une ambos grupos (también se puede usar el operador | ej: lenguajes_a | lenguajes_b)
todos_los_lenguajes = lenguajes_a.union(lenguajes_b)
print("3. Unión de sets (mezcla sin repetir):", todos_los_lenguajes)


# 4. .intersection() (ENCONTRAR COINCIDENCIAS)
# Extrae únicamente los elementos que se repiten exactamente en ambos conjuntos.
# (También se puede usar el operador & ej: lenguajes_a & lenguajes_b)
elementos_compartidos = lenguajes_a.intersection(lenguajes_b)
print("4. Intersección (lo que tienen en común):", elementos_compartidos)  # {'SQL'}


# 5. .difference() (ENCONTRAR DIFERENCIAS)
# Muestra los elementos que están en el primer conjunto pero que NO están en el segundo.
# (También se puede usar el operador - ej: lenguajes_a - lenguajes_b)
exclusivos_de_a = lenguajes_a.difference(lenguajes_b)
print("5. Diferencia (lo exclusivo del primer set):", exclusivos_de_a)  # {'Python'}
````
