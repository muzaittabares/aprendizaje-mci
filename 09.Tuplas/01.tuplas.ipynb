## 1. Reglas de Oro y Rendimiento (¿Por qué no usar simplemente listas?)

* **Inmutables:** Una vez creada, **NO puedes agregar, eliminar ni cambiar** sus elementos. (Esto garantiza que los datos no sean alterados por accidente).
* **Tienen orden:** A diferencia de los sets, sí conservan el orden exacto en el que fueron creadas y puedes acceder por índice.
* **Permiten duplicados:** Puedes tener el mismo valor repetido 100 veces.
* **Son "Hashables" (Casi siempre):** A diferencia de las listas, las tuplas **SÍ pueden usarse como claves en un diccionario** (siempre y cuando los elementos dentro de la tupla también sean inmutables).
* **Rendimiento:** Son más rápidas y ocupan menos memoria que las listas.

| Característica | Lista (`list`) | Tupla (`tuple`) |
| --- | --- | --- |
| **Mutabilidad** | Cambia (Mutable) | **Fija (Inmutable)** |
| **Tamaño en Memoria** | Mayor (Reserva espacio extra) | **Menor (Tamaño exacto)** |
| **Uso como Clave de Diccionario** | ❌ No permitido | **✅ Sí (si sus datos son inmutables)** |

## 2. Creación y la "Trampa de la Coma"

Lo que realmente crea una tupla en Python **es la coma, no los paréntesis** (los paréntesis solo se usan para agrupar o dar claridad).

```python
# Creación básica
t1 = (1, 2, 3)
t2 = 1, 2, 3      # Packing: ¡También es una tupla! (Sin paréntesis)
t_vacia = ()      # La única forma de crear una tupla vacía

# ⚠️ LA TRAMPA DE EXAMEN (Tupla de un solo elemento)
falsa_tupla = (5)   # Esto es un ENTERO (int) matemático normal
tupla_real = (5,)   # CORRECTO: ¡La coma es obligatoria para un solo elemento!
tupla_real2 = 5,    # También es válido

# Convertir otros iterables
t_texto = tuple("hola")      # ('h', 'o', 'l', 'a')
t_lista = tuple([1, 2, 3])   # (1, 2, 3)

```

## 3. Acceso, Slicing y "Modificación"

Accedes a ellas exactamente igual que a las listas. Y aunque no puedes modificarlas, puedes **concatenarlas** para crear tuplas nuevas.

```python
t = (10, 20, 30, 40, 50)

# ACCESO Y SLICING (Rebanado)
t[0]       # 10
t[-1]      # 50 (Último elemento)
t[1:4]     # (20, 30, 40)
t[::-1]    # (50, 40, 30, 20, 10) -> Tupla invertida (Crea una nueva)

# "MODIFICACIÓN" (Creando nuevas tuplas)
# t[0] = 99  -> ERROR (TypeError)
t2 = t + (60, 70)  # Concatenación: (10, ..., 50, 60, 70)
t3 = (1, 2) * 3    # Multiplicación: (1, 2, 1, 2, 1, 2)

```

## 4. Los Únicos 2 Métodos

Como las tuplas no pueden mutar, no tienen `.append()`, `.remove()`, ni `.sort()`. Solo tienen **dos** métodos para inspeccionar datos:

```python
t = (1, 5, 5, 5, 2, 8, 5)

print(t.count(5))     # 4 -> ¿Cuántas veces aparece el 5?
print(t.index(8))     # 5 -> ¿En qué índice/posición está el PRIMER 8 que encuentre?
# t.index(99) -> Dará ValueError si no existe

```

## 5. El Superpoder: Desempaquetado (Unpacking)

El desempaquetado es la razón número uno por la que verás tuplas en código avanzado. Te permite extraer los valores en variables separadas al instante.

```python
coordenadas = (4.5, -9.2)
x, y = coordenadas        # x = 4.5, y = -9.2

# ⚠️ TRAMPA: Si hay más elementos que variables, da ValueError.
# a, b = (1, 2, 3) -> ERROR

# 🌟 TRUCO PRO: Desempaquetado extendido (con asterisco)
t = (1, 2, 3, 4, 5)
primero, *medio, ultimo = t
# primero = 1
# medio   = [2, 3, 4]  (¡Ojo! El asterisco captura en formato de LISTA)
# ultimo  = 5

```

## 6. La Inmutabilidad "Engañosa" (Patrón de Entrevistas)

¿Una tupla puede cambiar? **La tupla no, pero los objetos mutables que vivan dentro de ella, SÍ.** Esta es la trampa teórica más famosa de Python.

```python
# Tenemos una tupla que contiene un número y una LISTA
trampa = (1, [2, 3])

# trampa[0] = 9        -> ERROR (No puedes cambiar a dónde apunta el índice 0)
# trampa[1] = [5, 6]   -> ERROR (No puedes cambiar a dónde apunta el índice 1)

# PERO... la lista en sí misma sigue siendo mutable:
trampa[1].append(4)    
print(trampa) # (1, [2, 3, 4]) -> ¡El interior de la tupla cambió!

# Esto significa que esta tupla NO puede ser clave de un diccionario (ya no es 100% hashable).

```

## 7. Tuplas con Nombre (Namedtuples) - Nivel Intermedio

Cuando las tuplas empiezan a tener 4 o más elementos, usar `t[3]` se vuelve confuso. Las `namedtuple` actúan como clases miniatura (sin métodos), fusionando la eficiencia de una tupla con la claridad de un diccionario.

```python
from collections import namedtuple

# 1. Definimos la plantilla (Nombre, [campos])
Punto = namedtuple('Punto', ['x', 'y', 'z'])

# 2. Creamos la tupla
p = Punto(10, 20, z=30)

# 3. Acceso híbrido
print(p[0])   # 10 (Funciona como tupla)
print(p.y)    # 20 (Funciona como objeto/propiedad -> Mucho más legible)

```

## 💡 8. Trucos Pro y Patrones de Ejercicios Clásicos

**1. Intercambio de variables (Swap) instantáneo:**
En otros lenguajes necesitas una variable temporal. En Python, las tuplas lo resuelven en una línea gracias al empaquetado/desempaquetado:

```python
a = 10
b = 20

a, b = b, a  # a vale 20, b vale 10

```

**2. Retornar múltiples valores en una función:**
Cuando usas `return x, y` en una función, Python está empaquetando esos valores en una tupla silenciosamente.

```python
def division_y_residuo(dividendo, divisor):
    return dividendo // divisor, dividendo % divisor  # Devuelve una tupla

div, res = division_y_residuo(10, 3) # div = 3, res = 1

```

**3. Usar tuplas compuestas como claves de diccionario:**
Clásico en algoritmos de mapas, laberintos o tableros de ajedrez donde necesitas guardar información sobre un punto $(x, y)$.

```python
# Un tablero guardado en un diccionario
tablero = {
    (0, 0): "Torre Blanca",
    (0, 1): "Caballo Blanco",
    (7, 7): "Torre Negra"
}

# ¿Qué hay en la posición x=0, y=1?
print(tablero[(0, 1)]) # "Caballo Blanco"

```
