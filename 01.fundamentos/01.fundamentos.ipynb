### Operadores Aritméticos en Python

Los operadores aritméticos son símbolos especiales que le indican a la computadora que realice operaciones matemáticas básicas y avanzadas entre uno o más valores.

### Tabla de Operadores Principales

| Operador | Nombre | Descripción | Ejemplo | Resultado |
| :---: | :--- | :--- | :--- | :--- |
| **`+`** | Suma | Suma dos valores | `5 + 3` | `8` |
| **`-`** | Resta | Resta el segundo valor del primero | `10 - 4` | `6` |
| **`*`** | Multiplicación | Multiplica ambos valores | `4 * 3` | `12` |
| **`/`** | División | Divide el numerador entre el denominador (decimal) | `7 / 2` | `3.5` |
| **`//`** | División entera | Divide y devuelve solo la parte entera (redondea hacia abajo) | `7 // 2` | `3` |
| **`%`** | Módulo (Residuos) | Devuelve el residuo de una división entera | `7 % 2` | `1` |
| **`**`** | Exponenciación | Eleva un número a la potencia de otro | `2 ** 3` | `8` |

Para sacar la raiz, solo utilizar exponenciacion ** .5

````python
suma = 2 + 2
resta = 10 - 8
multiplicacion = 3 * 3
division = 10 / 3
division_entera = 10 // 3
modulo = 7 % 2
exponenciacion = 2 ** 2
raiz = 9 ** .5
print([suma , resta , multiplicacion , division , division_entera , modulo , exponenciacion , raiz])
print(suma , resta , multiplicacion , division , division_entera , modulo , exponenciacion , raiz , sep="\n") #sep= irve para cambiar el carácter que separa los elementos al imprimirlos y el "\n genera un salto de linea"
````

````python
# 1. COMPLEX: Números con una parte imaginaria (llevan una 'j')
mi_complejo = 3 + 5j
print(mi_complejo)

# 2. ATAJOS (+=, -=, *=): Modifican la variable de forma rápida
puntos = 10
puntos += 5   # Es igual a decir: puntos = puntos + 5 (Ahora vale 15)
puntos *= 2   # Es igual a decir: puntos = puntos * 2 (Ahora vale 30)
print(puntos)

# 3. BITS (&, |, ^): Manipulan los "foquitos" internos de los números (0 y 1)
a = 5  # Binario: 101
b = 3  # Binario: 011

resultado_and = a & b   # 1, 0, 0 -> binario 001 -> normal 1 (ambos prendidos)
resultado_or = a | b    # 1, 1, 1 -> binario 111 -> normal 7 (con uno prendido)
resultado_xor = a ^ b   # 1, 1, 0 -> binario 110 -> normal 6 (solo si son diferentes)

print(resultado_and, resultado_or, resultado_xor)

# 4. IS: Comprueba si dos cosas son exactamente el mismo objeto en la memoria
lista_a = [1, 2, 3]
lista_b = [1, 2, 3]
son_exactamente_el_mismo = lista_a is lista_b  # Da False (tienen datos iguales, pero viven en cajones separados)
print(lista_a is lista_b)
````
