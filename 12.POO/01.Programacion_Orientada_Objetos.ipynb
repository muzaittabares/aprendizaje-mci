## 1. ¿Qué es la Programación Orientada a Objetos (POO)?

La **Programación Orientada a Objetos (POO)** es un paradigma de programación basado en el concepto de **"objetos"**. 
* Un objeto contiene tanto **datos** como **código**.
* Los datos se representan como **propiedades** o **atributos**.
* El código se representa como **métodos** (acciones que el objeto puede realizar).

La ventaja principal de la POO es el diseño modular y la reutilización de código mediante el principio **DRY** (*Don't Repeat Yourself* - No te repitas).

### Conceptos clave:
* **Clase:** Es el molde, plano o plantilla genérica.
* **Objeto:** Es la instancia física creada a partir de la clase (un ejemplar real).

---

## 2. Clases y Objetos en Python

Para definir una clase en Python se utiliza la palabra clave `class`. Por convención, se nombran usando la notación *CamelCase*.

```python
class Vehiculo:
    '''Clase base para representar cualquier vehículo'''
    pass

# Crear un objeto (instancia) de la clase
mi_auto = Vehiculo()
print(type(mi_auto))  # <class '__main__.Vehiculo'>

```

## 3. Constructores y Destructores

### Constructores (`__init__`)

Un constructor es un método especial que se ejecuta automáticamente al crear un objeto. Su propósito principal es **inicializar los atributos** (variables de instancia).

* Su nombre obligatorio es `__init__`.
* Recibe como primer parámetro `self`, que hace referencia al objeto actual que se está creando.

### Destructores (`__del__`)

Un destructor se utiliza para liberar recursos (como conexiones a bases de datos o archivos abiertos) justo antes de que el objeto sea destruido por el recolector de basura de Python.

```python
class ConexionBD:
    def __init__(self, servidor):
        self.servidor = servidor
        print(f"Conectado a {self.servidor}")

    def __del__(self):
        print(f"Cerrando conexión con {self.servidor}")

# Ejemplo de uso:
db = ConexionBD("Servidor_Principal")
del db  # Forzamos la destrucción del objeto (o ocurre al finalizar el programa)

```

## 4. Tipos de Variables en una Clase

Dentro de una clase, las variables se dividen en dos categorías principales:

1. **Variables de instancia:** Pertenecen exclusivamente a cada objeto. Se definen dentro del constructor `__init__` utilizando `self.variable`.
2. **Variables de clase:** Son compartidas por *todas* las instancias de la clase. Se declaran directamente dentro del cuerpo de la clase (fuera de cualquier método).

```python
class Empleado:
    # Variable de clase (compartida)
    empresa = "TechCorp"

    def __init__(self, nombre, salario):
        # Variables de instancia (únicas por objeto)
        self.name = nombre
        self.salary = salario

emp1 = Empleado("Ana", 15000)
emp2 = Empleado("Carlos", 18000)

print(emp1.empresa)  # TechCorp
print(emp2.empresa)  # TechCorp

```

*Nota:* Es posible **añadir, modificar o eliminar dinámicamente** variables o métodos de un objeto en tiempo de ejecución utilizando la notación de punto (`objeto.nuevo_attr = valor`) o la función `delattr()`.

## 5. Tipos de Métodos en Python

Existen tres tipos principales de métodos en una clase, cada uno con un propósito específico:

### A. Métodos de Instancia

* Operan sobre los datos del objeto (utilizan variables de instancia mediante `self`).
* Pueden acceder tanto a los atributos de instancia como a los de clase.

### B. Métodos de Clase (`@classmethod`)

* Operan sobre el estado de la clase en su conjunto, no sobre un objeto específico.
* Reciben como primer argumento `cls` en lugar de `self`.
* Se definen utilizando el decorador `@classmethod`.

### C. Métodos Estáticos (`@staticmethod`)

* Son funciones de utilidad general que se colocan dentro de la clase por motivos de organización.
* **No** reciben ni `self` ni `cls`. No pueden acceder ni modificar atributos de instancia ni de clase.
* Se definen utilizando el decorador `@staticmethod`.

```python
class Matematicas:
    factor_iva = 1.16  # Variable de clase

    def __init__(self, valor):
        self.valor = valor  # Variable de instancia

    # 1. Método de instancia
    2def calcular_total(self):
        return self.valor * Matematicas.factor_iva

    # 2. Método de clase
    @classmethod
    def cambiar_iva(cls, nuevo_iva):
        cls.factor_iva = nuevo_iva

    # 3. Método estático
    @staticmethod
    def es_positivo(numero):
        return numero > 0

# Pruebas:
print(Matematicas.es_positivo(-5))  # False
Matematicas.cambiar_iva(1.20)

```

## 6. Encapsulamiento y Ocultamiento de Datos

El **encapsulamiento** protege los datos internos de modificaciones accidentales o no autorizadas. En Python no existen los modificadores estrictos (`public`, `private`, `protected`), pero se simulan mediante convenciones de nombres basadas en guiones bajos:

* **Públicos:** Sin guiones (ej. `self.name`). Accesibles desde cualquier lugar.
* **Protegidos:** Un guion bajo (ej. `self._salario`). Indicador de uso interno (convención).
* **Privados:** Dos guiones bajos (ej. `self.__password`). Python aplica un mecanismo llamado *Name Mangling* para ocultarlos y evitar su acceso directo desde el exterior.

```python
class CuentaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular        # Público
        self.__saldo = saldo          # Privado

    def mostrar_saldo(self):          # Método público para acceder al dato protegido
        return self.__saldo

cuenta = CuentaBancaria("Luis", 5000)
print(cuenta.titular)                 # Funciona
# print(cuenta.__saldo)               # Error: AttributeError
print(cuenta.mostrar_saldo())         # Funciona correctamente (5000)

```

## 7. Herencia y Orden de Resolución de Métodos (MRO)

La **herencia** permite crear una clase nueva (hija/derivada) a partir de una clase existente (padre/base), heredando sus atributos y métodos para fomentar la reutilización.

```python
# Clase Base (Padre)
class Vehiculo:
    def __init__(self, marca, precio):
        self.marca = marca
        self.precio = precio

    def descripcion(self):
        print(f"Vehículo marca: {self.marca}, Precio: ${self.precio}")

# Clase Derivada (Hija)
class Auto(Vehiculo):
    def tocar_claxon(self):
        print(f"¡El {self.marca} está tocando la bocina! Beep beep.")

# Uso:
mi_auto = Auto("Tesla", 45000)
mi_auto.descripcion()      # Heredado del padre
mi_auto.tocar_claxon()     # Método propio del hijo

```

### Método de Resolución de Métodos (MRO - *Method Resolution Order*)

Cuando se trabaja con herencia múltiple (una clase hereda de varias clases a la vez), Python utiliza el orden MRO para decidir qué método ejecutar primero. Puedes consultar el orden MRO de cualquier clase llamando a `NombreClase.__mro__` o `NombreClase.mro()`.

## 8. Polimorfismo

El **polimorfismo** significa "muchas formas". Permite que diferentes objetos respondan al mismo nombre de método de formas distintas según su propia implementación.

```python
class Perro:
    def hacer_sonido(self):
        print("Guau guau")

class Gato:
    def hacer_sonido(self):
        print("Miau")

def hacer_ruido_animal(animal):
    # Función polimórfica: no le importa qué animal sea, siempre que tenga 'hacer_sonido'
    animal.hacer_sonido()

# Uso:
p = Perro()
g = Gato()

hacer_ruido_animal(p)  # Imprime: Guau guau
hacer_ruido_animal(g)  # Imprime: Miau

```

## Ejemplo Integral de Repaso

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.__edad = edad  # Privado

    def obtener_edad(self):
        return self.__edad

class Estudiante(Persona):
    def __init__(self, nombre, edad, carrera):
        super().__init__(nombre, edad)
        self.carrera = carrera

    def estudiar(self):
        print(f"{self.nombre} está estudiando {self.carrera}.")

# Instancia y ejecución
est = Estudiante("Sofía", 20, "Economía")
est.estudiar()
print(f"Edad (privada protegida): {est.obtener_edad()}")

```
