## 1. Los booleanos son técnicamente números (`bool` como subtipo de `int`)

En Python, el tipo de dato booleano es en realidad un subtipo de los números enteros (`int`). Esto significa por dentro que `True` equivale al número `1` y `False` equivale al número `0`. Se pueden sumar y operar matemáticamente, lo cual es muy común en preguntas de exámenes de programación.

```python
# Comprobamos que True vale 1 y False vale 0 por dentro
print(True == 1)  # Imprime: True
print(False == 0)  # Imprime: True

# ¡Los booleanos se pueden sumar entre sí!
print(True + True)  # Imprime: 2 (1 + 1)

# Se pueden mezclar en operaciones matemáticas con números
print(True + 5)  # Imprime: 6 (1 + 5)
print(False * 10)  # Imprime: 0 (0 * 10)

### 2. Valores "Truthy" y "Falsy"
No solo los valores explícitos True y False son evaluables; Python puede interpretar cualquier valor como un booleano dentro de una condición.

Falsy (equivalen a False): El número 0, cadenas vacías "", listas vacías [], diccionarios vacíos {} y el valor especial None.

Truthy (equivalen a True): Todo lo demás (números diferentes de cero, textos con contenido, listas con elementos). Esto se usa mucho en código real para verificar si una estructura tiene datos de forma abreviada (ej. if mi_lista:).

````python
# Comprobando el equivalente booleano con la función bool()
print(bool(0))  # Imprime: False (es un número cero)
print(bool(""))  # Imprime: False (texto vacío)
print(bool("hola"))  # Imprime: True (texto con contenido)
print(bool([]))  # Imprime: False (lista vacía)
print(bool([1, 2]))  # Imprime: True (lista con elementos)

# Aplicación real en condicionales:
mi_lista = ["manzana", "pera"]

if mi_lista:  # Python evalúa esto como True porque la lista no está vacía
  print("La lista tiene elementos")
````

### 3. Comparaciones encadenadas
Python permite escribir comparaciones dobles o encadenadas en una sola línea de código de manera muy natural, algo que en la mayoría de otros lenguajes de programación requiere obligatoriamente usar operadores lógicos combinados.

````python
edad = 20

# Esta comparación elegante...
print(1 < edad < 65)  # Imprime: True

# ...equivale exactamente a hacer esto usando 'and':
print((1 < edad) and (edad < 65))  # Imprime: True
````

### 4. Precedencia de operadores lógicos
Cuando combinas varios operadores lógicos en una misma línea, Python sigue un orden jerárquico estricto para evaluarlos: primero evalúa not, después evalúa and, y por último evalúa or. Si quieres evitar confusiones o errores de lógica, el consejo práctico es usar paréntesis para indicar qué operación debe hacerse primero.

````python
# El operador 'and' se ejecuta antes que el 'or'
# Es decir, primero evalúa (False and False) -> False
# Y luego evalúa (True or False) -> True
print(True or False and False)  # Imprime: True

# Para evitar confusiones y forzar el orden que tú quieras, usa paréntesis:
print((True or False) and False)  # Imprime: False
````

### 5. Evaluación de cortocircuito (Short-circuit)
Los operadores and y or a veces no evalúan la segunda parte de una operación si el resultado ya se puede determinar con la primera. Por ejemplo, en un and, si la primera parte es False, Python ya sabe que todo va a dar falso y ni siquiera se molesta en leer o ejecutar lo que sigue.

````python
def es_true():
  print("¡Me ejecuté!")
  return True


# Como el primer valor es False, Python hace un "cortocircuito"
# y nunca llama a la función 'es_true()'
print(False and es_true())  # Imprime solo: False (nunca verás el mensaje de la función)
````

### 6. Detalle de estilo: El uso correcto de is None
Para verificar si una variable contiene un valor nulo (None), la forma correcta y considerada "pythónica" en el estándar de programación de Python es utilizar la palabra reservada is en lugar del signo de igualdad ==.

````python
x = None

# Forma correcta y recomendada en Python:
print(x is None)  # Imprime: True

# Forma incorrecta (aunque a veces funcione, va contra las buenas prácticas):
print(x == None)  # Imprime: True
````

### Operadores Relacionales en Python

A continuación se muestra la tabla completa con los operadores relacionales, su descripción y ejemplos prácticos listos para probar en Python:

| Operador | Descripción | Ejemplo en Código | Resultado |
| :---: | :--- | :--- | :--- |
| `==` | Igual que | `print(5 == 69)` | `False` |
| `!=` | Diferente de | `print(5 != 3)` | `True` |
| `<` | Menor que | `print(6 < 20)` | `True` |
| `>` | Mayor que | `print(100 > 1)` | `True` |
| `<=` | Menor o igual que | `print(90 <= 90)` | `True` |
| `>=` | Mayor o igual que | `print(100 >= 101)` | `False` |
