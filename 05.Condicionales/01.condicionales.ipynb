### Estructuras Condicionales en Python (`if`, `elif`, `else`)

Las estructuras condicionales permiten que tus programas tomen decisiones basadas en condiciones (valores booleanos `True` o `False`). Esto cambia el flujo de ejecución del código dependiendo de lo que esté pasando en tiempo real.

---

## 1. La importancia de la indentación (Tabulaciones)
En Python, los bloques de código que pertenecen a un condicional **no van entre llaves `{}`**, sino que se definen estrictamente mediante la **indentación** (espacios o tabulación a la izquierda). Si no dejas el espacio correcto, Python arrojará un error de sintaxis (`IndentationError`).

---

## 2. Condicional Simple (`if`)
El bloque `if` (si...) evalúa una condición. Si el resultado es `True`, se ejecuta el código que está dentro de su bloque. Si es `False`, el código se ignora por completo.

```python
edad = 18

# Si la condición es verdadera, se ejecuta el bloque interno
if edad >= 18:
  print("Eres mayor de edad.")



## 3. `else` — El "si no"
El bloque `else` se ejecuta de forma automática cuando la condición del `if` principal resulta ser `False`. Es el plan de respaldo de tu estructura.

```python
edad = 15

if edad >= 18:
  print("Eres mayor de edad.")
else:
  print("Eres menor de edad.")

### 4. elif — Múltiples condiciones en cadena
Permite evaluar varias opciones en orden estricto. Python recorre las condiciones de arriba hacia abajo y se detiene en la primera que sea True (las demás ya ni siquiera se revisan).

Nota: Puedes incluir la cantidad de elif que necesites, pero solo puede haber un if al inicio y un else opcional al final.

````python
calificacion = 8

if calificacion >= 9:
  print("Excelente")
elif calificacion >= 7:
  print("Bueno")
elif calificacion >= 6:
  print("Suficiente")
else:
  print("Reprobado")
````

### 5. Condicionales anidados (if dentro de otro if)
Ocurre cuando colocas una estructura condicional dentro de otra. Aunque es útil para reglas complejas, anidar demasiados niveles dificulta la lectura del código (muchas veces es mejor combinarlas usando el operador and).

````python
edad = 20
tiene_id = True

if edad >= 18:
  if tiene_id:
    print("Puede entrar")
  else:
    print("Necesita identificación")
else:
  print("No puede entrar")
````

### 6. Conexión directa con Booleanos y Truthy/Falsy
Evita comparar explícitamente con True o False. Lo más "pythónico" y limpio es aprovechar el comportamiento Truthy/Falsy de las variables de forma directa.

````python
lista = [1, 2, 3]

# Forma redundante y que debes evitar:
if len(lista) > 0:
  print("tiene elementos")

# Forma correcta, limpia y preferida:
if lista:
  print("tiene elementos")

# Nota: Nunca escribas 'if variable == True:', con 'if variable:' basta.
````

### 7. El error clásico: = vs == dentro del if
Un error típico de principiante es utilizar un solo signo de igual = (que sirve para asignar valores) en lugar del doble igual == (que sirve para comparar). Python arrojará un error de sintaxis si intentas hacerlo.

````python
edad = 18

# if edad = 18:   # ERROR: Esto marcará un fallo de sintaxis en Python
#     print("...")

if edad == 18:  # Correcto: Esto compara de manera adecuada
  print("...")
````

### 8. Operador ternario (Condicional en una sola línea)
Es una forma compacta de escribir un bloque if/else sencillo cuando solo necesitas asignar un valor rápido a una variable.

````python
edad = 20

# Sintaxis: variable = [resultado_si_verdadero] if [condicion] else [resultado_falso]
estado = "mayor" if edad >= 18 else "menor"

print(estado)  # Imprime: mayor
````

### 9. Usar in dentro de condicionales
Puedes comprobar de forma muy elegante si un elemento se encuentra dentro de una colección (como una lista de textos o números) usando la palabra clave in.

````python
dia = "sábado"

if dia in ["sábado", "domingo"]:
  print("Es fin de semana")
````

### 10. Alternativa moderna: Estructura `match-case`

A partir de Python 3.10, se introdujo la estructura `match-case`. Funciona como una alternativa mucho más limpia y elegante cuando tienes que evaluar una variable contra múltiples valores posibles (similar al *switch* que existe en otros lenguajes de programación), evitando tener que escribir cadenas largas de `elif`.

```python
# Ejemplo utilizando match-case para evaluar un código de estado HTTP
codigo_estado = 200

match codigo_estado:
  case 200:
    print("Éxito: La solicitud se completó correctamente.")
  case 404:
    print("Error: No se encontró la página solicitada.")
  case 500:
    print("Error interno en el servidor.")
  case _:
    # El guion bajo (_) actúa como caso por defecto (equivalente al 'else')
    print("Código de estado desconocido.")

# Entendiendo `match-case` vs `if / elif / else`

A simple vista, tanto `match-case` como `if / elif / else` hacen lo mismo: **tomar decisiones según el valor de una variable**. 

La diferencia principal no es *lo que hacen*, sino **cómo lo hacen y para qué escenario fueron diseñados**.

---

## La analogía fácil: La cerradura vs. El guardia de seguridad

Imagina que tienes una llave y quieres saber qué puerta abre:

* **Escribir con `if / elif / else` es como un Guardia de Seguridad:**  
  El guardia va puerta por puerta haciendo preguntas complejas:  
  *"¿Tu llave es roja? No. ¿Tu llave es de metal Y mide más de 5 cm? No. ¿Tu llave es la número 200? ¡Sí!"*  
  El `if` es perfecto para evaluar condiciones variadas, rangos (`<`, `>`), operaciones lógicas (`and`, `or`) y preguntas complejas.

* **Escribir con `match-case` es como un Monedero / Clasificador de monedas:**  
  No hace preguntas complejas. Solo toma un objeto (el código `200`) y lo deja caer directamente en la ranura que le corresponde exactamente (`case 200`).  
  Es un **emparejamiento directo de patrones**.

# Guía Comparativa: `if / elif / else` vs `match-case`

A continuación tienes una tabla detallada con explicaciones claras, escenarios reales de uso y código de ejemplo para saber exactamente cuál elegir en cada situación:

| Criterio | `if / elif / else` | `match-case` |
| :--- | :--- | :--- |
| **¿Cuándo usarlo?** | Cuando necesitas evaluar **condiciones complejas, rangos numéricos o variables distintas** en cada paso. | Cuando vas a comparar **una sola variable contra varios valores exactos y fijos** (como un menú de opciones o códigos). |
| **Tipo de evaluación** | Funciona con comparaciones dinámicas: mayor que (`>`), menor que (`<`), intervalos (`18 <= edad <= 65`) y lógica con `and` / `or`. | Funciona con coincidencia exacta de valores o patrones fijos (`case 200`, `case "guardar"`). |
| **Mantenimiento y Lectura** | Ideal si hay pocos casos (2 a 4). Si hay demasiados `elif`, el código se vuelve largo y difícil de leer. | Extremadamente limpio y ordenado cuando hay muchos casos posibles (5 o más opciones concretas). |
| **Compatibilidad** | Compatible con **todas** las versiones de Python (desde la 1.0). | Disponible únicamente a partir de **Python 3.10** en adelante. |
| **Ejemplo de Código** | ```python<br># Evaluando un rango numérico<br>if nota >= 90:<br>    print("Sobresaliente")<br>elif nota >= 70:<br>    print("Aprobado")<br>else:<br>    print("Reprobado")<br>``` | ```python<br># Evaluando una opción fija de menú<br>match opcion:<br>    case "1":<br>        print("Iniciar juego")<br>    case "2":<br>        print("Cargar partida")<br>    case _:<br>        print("Opción inválida")<br>``` |

---

## Regla de Oro Práctica

* **Usa `if / elif / else`** si en tu mente la regla suena como: *"Si la edad es mayor a 18 Y tiene pase..."* (requiere cálculos, rangos o combinaciones).
* **Usa `match-case`** si en tu mente la regla suena como: *"Si la opción seleccionada es 'A', haz esto; si es 'B', haz aquello..."* (selección simple dentro de una lista fija).
