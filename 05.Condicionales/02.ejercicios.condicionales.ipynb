### **Ejercicio 1**

Crear un programa que pida al usuario una letra, y si es vocal, muestre el mensaje "Es vocal". Sino, decirle al usuario que no es vocal

```python
letra = (input("Escriba una letra: ")).lower()
vocal = "aeiou"

if not letra:
    print("Error, no introduciste nada")
elif len(letra) > 1:
    print("Escribiste mas de una letra")
elif letra not in vocal:
    print("Eso no es una vocal")
else:
    print("Eso es una vocal")
```

### **Ejercicio 2**

Escribir un programa que, dado un número entero, muestre su valor absoluto. Nota: para los números positivos su valor absoluto es igual al número (el valor absoluto de 52 es 52), mientras que, para los negativos, su valor absoluto es el número multiplicado por -1 (el valor absoluto de -52 es 52).

```python
numero = int(input("Escribe el numero del que deseas conocer su valor absoluto: "))

if numero >= 0:
    print(f"El valor absolutode tu numero es {numero}")
else:
    print(f"El valor absolutode tu numero es {numero * -1}")
```

```python
numero = int(input("Escriba el número del que desea conocer su valor absoluto: "))

print(f"El valor absoluto de tu número es {abs(numero)}")
```

### **Ejercicio 3**

Escribe un programa que pida dos palabras y diga si riman o no. Si coinciden las tres últimas letras tiene que decir que riman. Si coinciden sólo las dos últimas tiene que decir que riman un poco y si no, que no riman.

```python
palabra_1 = input("Escribe una palabra: ").lower()
palabra_2 = input("Escribe otra palabra: ").lower()

if len(palabra_1) >= 3 and len(palabra_2) >= 3:
    if palabra_1[-3:] == palabra_2[-3:]:
        print("Tus palabras riman")
    elif palabra_1[-2:] == palabra_2[-2:]:
        print("Riman un poco")
    else:
        print("No riman")
else:
    print("Error, una de tus palabras tiene menos de 3 letras")
```

### **Ejercicio 4**

Crear un programa que permita al usuario elegir un candidato por el cual votar. Las posibilidades son: candidato A por el partido rojo, candidato B por el partido verde, candidato C por el partido azul. Según el candidato elegido (A, B ó C) se le debe imprimir el mensaje “Usted ha votado por el partido [color que corresponda al candidato elegido]”. Si el usuario ingresa una opción que no corresponde a ninguno de los candidatos disponibles, indicar “Opción errónea”.

```python
a = "Partido rojo"
b = "Partido verde"
c = "Partido azul"

election = input("Esriba A si desea votar por el partido rojo\nEscriba B si desea votar por el partido verde\nEscriba C si desea votar por el partido azul: ").lower()

match election: 
    case "a":
        print(f"Usted ha votado por el {a}")
    case "b":
        print(f"Usted ha votado por el {b}")
    case "c":
        print(f"Usted ha votado por el {c}")
    case _:
        print("Error, esa opcion no existe")
```
