### 1. Creación y Características Principales

* **Claves Únicas e Inmutables:** Las claves no se pueden repetir y deben ser de tipos inmutables (`str`, `int`, `float`, `tuple`).
* **Valores Libres:** Los valores pueden ser de cualquier tipo (enteros, listas, otros diccionarios).
* **Ordenados:** Desde Python 3.7+, mantienen el orden de inserción de las claves.

```python
# Crear diccionarios vacíos
dict_vacio_1 = {}
dict_vacio_2 = dict()

# Diccionario con diversos tipos de datos
persona = {
    "nombre": "Ana",
    "edad": 25,
    "es_estudiante": True,
    "habilidades": ["Python", "SQL"]
}

```

### 2. Acceso y Modificación de Elementos

* **Con corchetes `d[clave]`:** Lee o crea/actualiza un valor. Lanza `KeyError` si la clave no existe.
* **Con `.get(clave, valor_default)`:** Lee de forma segura. Si la clave no existe, retorna `None` o el valor por defecto sin lanzar error.

```python
usuario = {"nombre": "Carlos", "rol": "admin"}

# Lectura directa y segura
print(usuario["nombre"])         # Carlos
print(usuario.get("edad", 18))   # 18 (No existe 'edad', usa el default)

# Agregar y modificar
usuario["edad"] = 30             # Agrega la clave 'edad'
usuario["rol"] = "superadmin"    # Modifica la clave 'rol'

```

---

### 3. Eliminación de Elementos

* `.pop(clave)`: Elimina la clave y **retorna** su valor.
* `.popitem()`: Elimina y retorna el **último** par `(clave, valor)` insertado.
* `del d[clave]`: Instrucción para eliminar un par por su clave.
* `.clear()`: Vacía completamente el diccionario.

```python
data = {"a": 1, "b": 2, "c": 3, "d": 4}

val = data.pop("b")       # Retorna 2 -> data queda {'a': 1, 'c': 3, 'd': 4}
ultimo = data.popitem()   # Retorna ('d', 4) -> data queda {'a': 1, 'c': 3}
del data["a"]             # data queda {'c': 3}
data.clear()              # data queda {}

```

### 4. Métodos Principales de Búsqueda e Iteración

#### A. Vistas de Claves, Valores e Items

* `.keys()`: Retorna todas las claves.
* `.values()`: Retorna todos los valores.
* `.items()`: Retorna tuplas `(clave, valor)`.
* Operador `in`: Verifica la existencia de una clave.

```python
producto = {"nombre": "Laptop", "precio": 1200, "stock": 10}

print("precio" in producto)  # True
print(producto.keys())       # dict_keys(['nombre', 'precio', 'stock'])
print(producto.values())     # dict_values(['Laptop', 1200, 10])

# Iterar sobre clave y valor simultáneamente
for clave, valor in producto.items():
    print(f"{clave}: {valor}")

```

#### B. Inicialización y Valores por Defecto

* `dict.fromkeys(secuencia, valor_default)`: Crea un diccionario asignando el mismo valor a cada clave.
* `.setdefault(clave, valor_default)`: Si la clave existe la devuelve; si no, la crea con el valor por defecto.

```python
# Crear plantilla
claves = ["math", "science", "english"]
notas = dict.fromkeys(claves, 0)  # {'math': 0, 'science': 0, 'english': 0}

# Garantizar lista dentro de clave
agenda = {}
agenda.setdefault("amigos", []).append("Luis")  # {'amigos': ['Luis']}

```

### 5. Combinación de Diccionarios (Merge)

* `.update(d2)`: Modifica el diccionario original agregando o sobrescribiendo los pares de `d2`.
* Operador `|` (Python 3.9+): Retorna un **nuevo** diccionario con la unión de ambos.

```python
d1 = {"a": 1, "b": 2}
d2 = {"b": 20, "c": 3}

# Unión con operador |
union = d1 | d2
print(union)  # {'a': 1, 'b': 20, 'c': 3}

# Modificación in-place con update
d1.update(d2)
print(d1)     # {'a': 1, 'b': 20, 'c': 3}

```

### 6. Funciones Integradas con Diccionarios

| Función | Descripción |
| --- | --- |
| `len(d)` | Retorna el número de pares clave-valor. |
| `max(d)` / `min(d)` | Retorna la clave máxima/mínima (por orden alfabético/numérico). |
| `max(d, key=d.get)` | Retorna la clave que contiene el valor máximo. |
| `sorted(d)` | Retorna una lista ordenada de las claves. |

```python
precios = {"manzana": 0.5, "banana": 0.2, "cereza": 1.5}

print(len(precios))                # 3
print(max(precios, key=precios.get)) # cereza (clave con el valor más alto)
print(sorted(precios))             # ['banana', 'cereza', 'manzana']

```

### 7. Diccionarios Anidados

Un diccionario puede guardar otros diccionarios para estructurar datos complejos. Se accede encadenando corchetes `[clave1][clave2]`.

```python
empresa = {
    "empleado1": {"nombre": "Ana", "puesto": "Dev"},
    "empleado2": {"nombre": "Luis", "puesto": "Diseñador"}
}

# Acceso en profundidad
print(empresa["empleado1"]["nombre"])  # Ana

# Modificación profunda
empresa["empleado2"]["puesto"] = "Lead Designer"

```

### 8. Creación de Diccionarios desde Listas (`zip`)

Permite emparejar dos listas (una de claves y otra de valores) para construir un diccionario directamente.

```python
claves = ["id", "nombre", "nivel"]
valores = [101, "Sofia", "Avanzado"]

usuario = dict(zip(claves, valores))
print(usuario)  # {'id': 101, 'nombre': 'Sofia', 'nivel': 'Avanzado'}

```

### 9. Desempaquetado (*Unpacking*) con `**`

El operador `**` extrae todos los pares clave-valor de un diccionario. Es útil para fusionar estructuras o pasar argumentos.

```python
base = {"host": "localhost", "port": 8080}
extra = {"user": "admin", "pass": "1234"}

# Fusionar en un nuevo diccionario
config = {**base, **extra, "timeout": 30}
print(config)  # {'host': 'localhost', 'port': 8080, 'user': 'admin', 'pass': '1234', 'timeout': 30}

```

### 10. Ordenamiento Avanzado por Valores

Para ordenar las claves de un diccionario según sus valores, se utiliza `sorted()` indicando que el criterio de evaluación sea el valor (`item[1]`).

```python
puntajes = {"Juan": 85, "Maria": 95, "Pedro": 70}

# Ordenar por valor de menor a mayor
ordenado_por_valor = dict(sorted(puntajes.items(), key=lambda item: item[1]))

print(ordenado_por_valor)  # {'Pedro': 70, 'Juan': 85, 'Maria': 95}

```

### 11. Copia de Diccionarios: Superficial (*Shallow*) vs Profunda (*Deep*)

1. **`.copy()` (Copia Superficial):** Duplica el diccionario principal. Si contiene listas o subdiccionarios, los cambios en las estructuras internas afectarán al original.
2. **`copy.deepcopy()` (Copia Profunda):** Genera una copia totalmente independiente, duplicando también las estructuras anidadas.

```python
import copy

original = {"a": 1, "lista": [10, 20]}

# Copia superficial
copia_superficial = original.copy()
copia_superficial["lista"].append(99)
print(original["lista"])  # [10, 20, 99] (¡El original cambió!)

# Copia profunda
original_2 = {"a": 1, "lista": [10, 20]}
copia_profunda = copy.deepcopy(original_2)
copia_profunda["lista"].append(99)
print(original_2["lista"])  # [10, 20] (Permaneció intacto)

```

### 12. Comprensión de Diccionarios (*Dict Comprehension*)

Sintaxis concisa para transformar o filtrar diccionarios en una sola línea.

```python
# Sintaxis: {clave_expr: valor_expr for elemento in iterable if condicion}

# Crear diccionario de cuadrados
cuadrados = {x: x**2 for x in range(1, 6)}
# Resultado: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Filtrar solo elementos con valor mayor a 50
temperaturas = {"Lunes": 20, "Martes": 55, "Miércoles": 60}
calurosos = {dia: temp for dia, temp in temperaturas.items() if temp > 50}
# Resultado: {'Martes': 55, 'Miércoles': 60}

```

| Método / Sintaxis | Propósito General |

| `d[k]` | Lee/Escribe el valor de la clave `k` (Lanza `KeyError` si no existe al leer). |

| `d.setdefault(k, def)` | Obtiene `k`. Si no existe, lo inserta con el valor `def` y lo retorna. |
| `d.update(d2)` | Inserta/Actualiza claves y valores desde otro diccionario o iterable `d2`. |
| `d1 | d2` | Retorna un nuevo dict fusionando `d1` y `d2` (Python 3.9+). |
| `del d[k]` | Elimina la clave `k` (Lanza `KeyError` si no existe). |
| `d.pop(k, def)` | Elimina `k` y **retorna** su valor (o `def` si no existe). |
| `d.popitem()` | Elimina y retorna el último par `(clave, valor)` insertado. |
| `d.clear()` | Vacía completamente el diccionario. |
| `d.keys()` | Retorna una vista iterable de todas las claves. |
| `d.values()` | Retorna una vista iterable de todos los valores. |
| `d.items()` | Retorna una vista iterable de tuplas `(clave, valor)`. |
| `dict.fromkeys(seq, def)` | Crea un dict con las claves de la secuencia `seq` y un valor fijo `def`. |
| `k in d` | Devuelve `True` si la clave `k` existe en el diccionario `d`. |
| `d.copy()` | Copia superficial (*shallow copy*) del diccionario. |
