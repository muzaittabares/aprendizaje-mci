## 1. Reglas de Oro y Rendimiento (Por qué usar Sets)

* **No hay duplicados:** Si intentas agregar un elemento que ya existe, Python lo ignora silenciosamente.
* **No tienen orden:** No puedes acceder a ellos con índices (ej. `mi_set[0]` dará error).
* **Los elementos deben ser inmutables (Hashables):** Puedes guardar números, strings o tuplas, pero **NO** puedes guardar listas, diccionarios ni otros sets dentro de un set.
* **Rendimiento (Big O):** El motivo principal para usar sets en algoritmos es la velocidad extrema.

| Operación | Lista (`list`) | Conjunto (`set`) |
| --- | --- | --- |
| Búsqueda (`x in ...`) | $O(n)$ | **$O(1)$** |
| Agregar elemento | $O(1)$ | **$O(1)$** |
| Eliminar elemento | $O(n)$ | **$O(1)$** |

## 3. Modificación (Agregar y Eliminar)

```python
s = {1, 2, 3}

# AGREGAR
s.add(4)          # {1, 2, 3, 4} (Agrega un solo elemento)

# ELIMINAR
s.remove(6)       # Elimina el 6. Si no existe, lanza un ERROR (KeyError).
s.discard(99)     # Elimina el 99. Si no existe, NO HACE NADA (Seguro).
s.pop()           # Elimina y devuelve un elemento al azar.
s.clear()         # Vacía el set por completo.

```

## 4. Matemáticas de Conjuntos (El núcleo de los ejercicios)

Para cada operación existe un **operador** (más rápido, exige que ambos sean sets) y un **método** (acepta cualquier iterable, como listas).

| Operación | Operador | Método | Resultado | Explicación |
| --- | --- | --- | --- | --- |
| **Unión** | `A | B` | `A.union(B)` | `{1, 2, 3, 4, 5}` | Une ambos (sin repetir). |
| **Intersección** | `A & B` | `A.intersection(B)` | `{3}` | Lo que está en AMBOS. |
| **Diferencia** | `A - B` | `A.difference(B)` | `{1, 2}` | En `A` pero NO en `B`. |
| **Dif. Simétrica** | `A ^ B` | `A.symmetric_difference(B)` | `{1, 2, 4, 5}` | En `A` o `B`, pero NO en ambos. |

```python
A = {1, 2, 3}
B = {3, 4, 5}

# Operaciones con múltiples sets a la vez
C = {5, 6, 7}
set.union(A, B, C)        # Une los tres de un jalón: {1, 2, 3, 4, 5, 6, 7}
set.intersection(A, B, C) # Intersección de los tres (útil si te dan una lista de sets)

```

## 5. Actualización en Sitio (Mutar el conjunto original)

Si necesitas guardar el resultado en el mismo conjunto para ahorrar memoria, usa las operaciones de actualización.

| Operación | Operador | Método |
| --- | --- | --- |
| **Unión en sitio** | `A |= B` | `A.update(B)` *(¡Trampa! No se llama union_update)* |
| **Intersección en sitio** | `A &= B` | `A.intersection_update(B)` |
| **Diferencia en sitio** | `A -= B` | `A.difference_update(B)` |
| **Dif. Simétrica en sitio** | `A ^= B` | `A.symmetric_difference_update(B)` |

## 6. Subconjuntos, Superconjuntos e Inspección

Es vital entender la diferencia entre un subconjunto *normal* (donde ambos pueden ser idénticos) y uno *estricto* (donde uno debe ser genuinamente más grande).

```python
A = {1, 2}
B = {1, 2, 3}

# NORMALES (Permiten que A y B sean iguales)
A <= B  # True (A es subconjunto de B)
A <= A  # True (A es subconjunto de sí mismo)
B >= A  # True (B es superconjunto de A)

# ESTRICTOS (No permiten igualdad)
A < B   # True (A está en B, pero B es más grande)
A < A   # False (Trampa común: un set no es subconjunto estricto de sí mismo)

# DISJUNTOS (No tienen absolutamente nada en común)
A.isdisjoint({9, 10}) # True

```

```python
lista = [3, 1, 2, 3, 1]

# Si NO importa el orden:
limpia_rapida = list(set(lista)) # [1, 2, 3] (Se pierde el orden original)

# Si SÍ importa el orden (El truco real para exámenes):
limpia_ordenada = list(dict.fromkeys(lista)) # [3, 1, 2]

```
