## 1. Regla de Oro: ¿Cuál elegir?

* **Usa `for**` cuando **sabes cuántas veces** vas a iterar (ej. recorrer una lista, un texto, o un rango de números). Python usa un enfoque "For-Each" (por cada elemento en...).
* **Usa `while**` cuando **NO sabes cuántas veces** iterarás, pero sabes que debe detenerse cuando se cumpla una **condición**.

---

## 2. El Bucle `for` y sus Herramientas (La navaja suiza)

### 🔹 Iterar Colecciones

```python
texto = "hola"
for letra in texto:
    print(letra)

```

### 🔹 `range()`: Cuando solo necesitas números

`range(inicio, fin, salto)` genera secuencias numéricas (el `fin` NUNCA se incluye).

```python
for i in range(5):          # 0, 1, 2, 3, 4
for i in range(2, 6):       # 2, 3, 4, 5
for i in range(0, 10, 2):   # 0, 2, 4, 6, 8 (Pares)

```

### 🔹 `reversed()` y `sorted()` combinados con `for`

Para cambiar el orden de iteración al vuelo sin modificar la lista original.

```python
for i in reversed(range(5)):     
    print(i) # 4, 3, 2, 1, 0

for palabra in sorted(["banana", "apple", "cherry"]):
    print(palabra)  # apple, banana, cherry

```

### 🔹 `enumerate()`: Iterar pidiendo el Índice

Si necesitas el elemento **Y** su posición.

```python
lista = ["a", "b", "c"]
for indice, valor in enumerate(lista):
    print(f"En el índice {indice} está {valor}")

```

### 🔹 `zip()`: Iterar varias listas al mismo tiempo

Une iterables. Se detiene cuando la lista más corta se acaba.

```python
nombres = ["Ana", "Bob"]
edades = [25, 30]
for nombre, edad in zip(nombres, edades):
    print(f"{nombre} tiene {edad} años")

```

### 🔹 Desempaquetado en `for` con tuplas anidadas

```python
puntos = [(1, 2), (3, 4), (5, 6)]
for x, y in puntos:
    print(f"x={x}, y={y}")

# Nivel avanzado: enumerate + tuplas al mismo tiempo
for i, (x, y) in enumerate(puntos):
    print(f"Punto {i}: ({x}, {y})")

```

## 3. El Bucle `while` y el Operador Morsa

El bloque de código se repite **mientras** la condición sea `True`.

```python
contador = 0
while contador < 5:
    print(contador)
    contador += 1   # ¡VITAL! Evita el bucle infinito

```

### 🔹 `while` con el operador morsa `:=` (Python 3.8+)

Evita repetir código asignando y evaluando en la misma línea. Muy común en ejercicios modernos.

```python
# Antes (dos líneas repetidas para el input)
dato = input("Escribe algo: ")
while dato != "salir":
    print(dato)
    dato = input("Escribe algo: ")

# Con operador morsa (una línea limpia)
while (dato := input("Escribe algo: ")) != "salir":
    print(dato)

```

## 4. Controladores de Flujo y Trampas de Examen

| Palabra Clave | ¿Qué hace? | Ejemplo de uso clásico |
| --- | --- | --- |
| **`break`** | **Destruye** el bucle por completo. Sale inmediatamente. | Detener la búsqueda en cuanto encuentras el dato. |
| **`continue`** | **Salta a la siguiente iteración**. Ignora el código de abajo. | Saltarse elementos que no cumplen un filtro (ej. impares). |
| **`pass`** | **No hace nada**. Es relleno estructural. | Dejar un bucle vacío temporalmente para programarlo después. |

### ⚠️ Trampa Clásica: Reasignar la variable del `for` no afecta la iteración

```python
for i in range(5):
    i = 100   # Esto NO afecta la siguiente vuelta del bucle
    print(i)  # Imprime 100, 100, 100, 100, 100
# Internamente, el bucle sigue pidiéndole el siguiente valor a range(), ignorando tu "100"

```

## 5. El Secreto Mejor Guardado: `else` en bucles

* **El `else` SE EJECUTA** si el bucle terminó naturalmente.
* **El `else` NO SE EJECUTA** si el bucle fue destruido por un `break`.

```python
buscar = 5
for num in [1, 2, 3]:
    if num == buscar:
        print("Encontrado")
        break
else:
    print("El número no estaba en la lista") 

```

## 6. Bucles Anidados y su Mayor Peligro

Un bucle dentro de otro. Por cada paso del exterior, el interior da todos sus pasos.

### ⚠️ Trampa Mortal: `break` solo rompe el bucle interno

Esta es la fuente #1 de bugs en ejercicios de matrices.

```python
for i in range(3):
    for j in range(3):
        if j == 1:
            break   # Solo rompe el bucle de "j", NO el de "i"
        print(i, j)
# El bucle de "i" sigue corriendo normalmente. 
# Solución para romper ambos: usar una bandera (flag), o meter la lógica en una función y hacer return.

```

## 7. Alternativas "Pythónicas": Comprehensions

En nivel intermedio/avanzado, rara vez usarás un `for` básico para crear listas. Las *Comprehensions* son obligatorias.

```python
# ❌ Bucle tradicional
cuadrados = []
for x in range(5):
    cuadrados.append(x**2)

# ✅ List Comprehension (La forma "pythónica")
cuadrados = [x**2 for x in range(5)]

# Con condición (Filtro)
pares = [x for x in range(10) if x % 2 == 0]

# Dict comprehension
cuadrados_dict = {x: x**2 for x in range(5)}

```

## 8. Herramientas Avanzadas: `itertools`

Para cuando `range` o `zip` no son suficientes. Especial para combinatoria.

```python
from itertools import combinations, permutations, product

list(combinations([1,2,3], 2))   # [(1,2), (1,3), (2,3)]
list(permutations([1,2,3], 2))   # [(1,2), (1,3), (2,1), (2,3), (3,1), (3,2)]
list(product([1,2], [3,4]))      # [(1,3), (1,4), (2,3), (2,4)]

```

**5. "Two Pointers" / Ventana Deslizante (Sliding Window):**
Clásico en ejercicios de algoritmos con arrays/strings usando `while`.

```python
# Ejemplo: verificar si un string es palíndromo
def es_palindromo(s):
    izq, der = 0, len(s) - 1
    while izq < der:
        if s[izq] != s[der]:
            return False
        izq += 1
        der -= 1
    return True

```

````python
def es_palindromo(s):
    # PREPARACIÓN: Colocamos nuestros dos "dedos" (punteros) en los extremos.
    
    # izq = 0 (Nuestro dedo izquierdo apunta a la primera posición)
    # len(s) es 5 (la cantidad total de letras). Al restarle 1, nos da 4.
    # der = 4 (Nuestro dedo derecho apunta a la última posición)
    izq, der = 0, len(s) - 1
    
    # INICIA EL BUCLE: La condición es "Mientras el dedo izquierdo no cruce al derecho".
    # VUELTA 1: ¿0 es menor que 4? Sí. Entramos al bucle.
    # VUELTA 2: ¿1 es menor que 3? Sí. Entramos al bucle.
    # VUELTA 3: ¿2 es menor que 2? Falso. El bucle se rompe y termina.
    while izq < der:
        
        # COMPARACIÓN: Verificamos si las letras que estamos tocando son DIFERENTES.
        # VUELTA 1: s[0] es 'r'. s[4] es 'r'. ¿Son diferentes? FALSO. (El if se ignora)
        # VUELTA 2: s[1] es 'a'. s[3] es 'a'. ¿Son diferentes? FALSO. (El if se ignora)
        if s[izq] != s[der]:
            # (Si estuviéramos leyendo "hola", en la vuelta 1 compararíamos 'h' y 'a'.
            # Como sí son diferentes, Python entraría aquí y escupiría False inmediatamente, 
            # matando la función sin seguir revisando).
            return False
        
        # MOVIMIENTO: Como las letras fueron iguales, acercamos los dedos un paso al centro.
        # VUELTA 1: izq pasa de 0 a 1. der pasa de 4 a 3. (Termina la vuelta 1, volvemos al while)
        # VUELTA 2: izq pasa de 1 a 2. der pasa de 3 a 2. (Termina la vuelta 2, volvemos al while)
        izq += 1
        der -= 1

    # VEREDICTO: Si el código llegó hasta esta línea, significa que el bucle terminó 
    # de revisar toda la palabra y NUNCA entró al 'if' que devuelve False.
    # Como nunca encontró letras diferentes, estamos 100% seguros de que es un palíndromo.
    return True

# --- Prueba del código ---
print(es_palindromo("radar"))  # Imprimirá True
````
