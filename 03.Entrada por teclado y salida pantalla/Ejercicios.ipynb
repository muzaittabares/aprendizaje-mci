### **Ejercicio 2**

Se desea tener un algoritmo que permita determinar y mostrar el promedio que ha obtenido un alumno en un determinado curso, conociendo las notas de: tres prácticas, el examen parcial y el examen final.

Considere:

PP = ( P1 + P2 +P3 ) / 3 PROM = ( PP + 2*EP + 3*EF ) / 6

Donde: P1, P2, P3 : Practicas

PP: promedio de práctica

PROM: promedio

EP: examen parcial

EF: examen final

```python
p1 = float(input("Escriba la calificaion obtenida en la primer practica: "))
p2 = float(input("Escriba la calificaion obtenida en la segunda practica: "))
p3 = float(input("Escriba la calificaion obtenida en la tercer practica: "))
ep = float(input("Escriba la calificacion obtenida en el examen parcial: "))
ef = float(input("Escriba la calificacion obtenida en el examen final: "))
promedio = ((p1 + p2 + p3) / 3 + (2 * ep) + (3 * ef)) / 6
print(f"Tu promedio final es de: {promedio}")
```

### **Ejercicio 3**

Escribir un programa que solicite al usuario un vocal en minuscula, y luego una letra en mayúsculas. El programa debe convertir la letra en minúscula y la vocal en mayúscula, y al final, deben ser concatenadas ambas

```python
vowel = input("Escriba una vocal: ")
letra = input("Escriba una letra: ")
vocales = "aeiou"

if vowel not in vocales:
    print("Eso no es una vocal")
elif vowel.isupper():
    print("Error, escribiste la vocal en Mayuscula, no en minuscula")
elif letra.islower():
    print("Error, escribiste la letra en Minuscula, no en mayuscula")
elif letra.lower() in vocales:
    print("Tiene que ser consonante")
else:
    print(vowel.upper() + letra.lower())
```

### Exercise 5. Display variables with a separator

Practice Problem: Use the print() function to format the words “Name”, “Is”, and “James” so they appear with three asterisks (***) between them.

Exercise Purpose: The print() function is more powerful than it looks. This exercise teaches the use of the sep (separator) keyword argument, which allows you to define how multiple items should be joined together in the output without manually typing strings.

Given Input:

Strings: “Name”, “Is”, “James”
Expected Output:

Name*** Is***James

```python
print("Name" , "Is" , "james" , sep="***")
```

### Exercise 6. Convert decimal number to octal
Practice Problem: Accept an integer from the user and display it as an octal number (Base 8).

Exercise Purpose: Computers often use different numbering systems (Hexadecimal, Octal, Binary). This exercise demonstrates how to use “Format Specifiers” in strings to automatically convert values between number systems for display.

Given Input: Decimal: 8

Expected Output:

The octal number of decimal number 8 is 10

```python
numero = int(input("Ingrese un número decimal: "))

# El :o dentro de {numero:o} le dice a Python que lo imprima en octal
print(f"The octal number of decimal number {numero} is {numero:o}")
```
