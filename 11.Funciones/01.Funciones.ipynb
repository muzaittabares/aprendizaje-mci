## 1. Regla de Oro y Anatomía Básica de una Función

### 💡 Regla de Oro: `return` vs. `print()`

* **`print()`** es una función de **salida visual para el usuario en consola**. No devuelve ningún dato utilizable por el programa; su valor de retorno es siempre `None`.
* **`return`** es la **salida de datos de la función hacia el programa**. Detiene inmediatamente la ejecución de la función y entrega un valor a quien la invocó para que pueda guardarse en una variable, reusarse o procesarse.

---

### 🔹 Sintaxis, Invocación y Flujo de Ejecución

Una función se define mediante la palabra clave `def`, seguida de su nombre, paréntesis `()` con los parámetros de entrada y dos puntos `:`.

```python
# 1. Definición de la función (Python la guarda en memoria, NO la ejecuta aún)
def calcular_area_rectangulo(base, altura):
    area = base * altura
    return area  # Devuelve el resultado y finaliza la función

# 2. Invocación / Llamada a la función
resultado = calcular_area_rectangulo(5, 3)  # Asigna el valor retornado (15)
print(f"El área es: {resultado}")  # Salida: El área es: 15

```

---

### ⚠️️ Trampa Clásica: Confundir `print()` con `return`

```python
def funcion_con_print(a, b):
    print(a + b)

def funcion_con_return(a, b):
    return a + b

# Intento de reutilizar los resultados:
x = funcion_con_print(2, 3)   # Muestra "5" en pantalla
y = funcion_con_return(2, 3)  # No muestra nada, pero guarda 5 en 'y'

print(x)  # Salida: None  (¡TRAMPA! No puedes operar con 'x')
print(y)  # Salida: 5     (Es un número válido)

# z = x + 10  # ❌ TypeError: unsupported operand type(s) for +: 'NoneType' and 'int'
z = y + 10    # ✅ Funciona perfectamente: z = 15

```

> **Nota:** Si una función finaliza su ejecución sin llegar a un enunciado `return` explícito, Python retornará automáticamente `None`.

### 🔹 Buenas Prácticas de Nombramiento (PEP 8) y Docstrings

Según la guía de estilo **PEP 8**:

* Usa **snake_case** para los nombres de funciones (`calcular_promedio`, `validar_usuario`).
* Utiliza verbos descriptivos que indiquen la acción realizada.
* Documenta la función usando **Docstrings** (`""" ... """`) justo debajo de la firma.

```python
def obtener_calificacion(promedio: float) -> str:
    """Calcula la letra de calificación correspondiente a un promedio numérico.

    Args:
        promedio (float): Puntaje numérico entre 0 y 100.

    Returns:
        str: Letra 'A', 'B', 'C' o 'F' según el rango.
    """
    if promedio >= 90:
        return "A"
    elif promedio >= 80:
        return "B"
    elif promedio >= 70:
        return "C"
    return "F"

# Acceder a la documentación en tiempo de ejecución:
print(obtener_calificacion.__doc__)

```

## 2. Parámetros y Argumentos (El Arte de Pasar Datos)

### 🔹 Posicionales vs. Keyword Arguments

```python
def crear_perfil(nombre, edad, profesion):
    return f"{nombre} tiene {edad} años y es {profesion}."

# Argumentos Posicionales (El orden importa strictly)
print(crear_perfil("Ana", 28, "Ingeniera"))

# Keyword Arguments / Por Nombre (El orden NO importa)
print(crear_perfil(profesion="Diseñadora", nombre="Carlos", edad=35))

# Mezcla (Los posicionales van SIEMPRE al inicio)
print(crear_perfil("Beatriz", profesion="Doctora", edad=40))

```

### 🔹 Valores por Defecto

Permiten invocar funciones sin especificar todos sus argumentos:

```python
def conectar_bd(host="localhost", puerto=5432, usuario="admin"):
    return f"Conectando a {host}:{puerto} con usuario '{usuario}'"

print(conectar_bd())                         # Usa todos los defaults
print(conectar_bd("192.168.1.1"))           # Cambia solo 'host'
print(conectar_bd(puerto=3306, usuario="root")) # Cambia puerto y usuario

```

### ⚠️ TRAMPA MORTAL: Argumentos Mutables por Defecto

Cuando usas un objeto mutable (como una lista `[]` o un diccionario `{}`) como valor por defecto, Python **crea el objeto una sola vez cuando define la función**, no en cada llamada.

```python
# ❌ CÓDIGO CON BUG (Comportamiento inesperado)
def agregar_item_incorrecto(item, lista=[]):
    lista.append(item)
    return lista

print(agregar_item_incorrecto("A")) # ['A']
print(agregar_item_incorrecto("B")) # ['A', 'B']  <-- ¡LA LISTA PERSISTE!
print(agregar_item_incorrecto("C")) # ['A', 'B', 'C']

```

#### ✅ Solución Pythónica Profesional: Utilizar `None`

```python
def agregar_item_correcto(item, lista=None):
    if lista is None:
        lista = []  # Se crea una nueva lista en CADA llamada si no se proporciona una
    lista.append(item)
    return lista

print(agregar_item_correcto("A")) # ['A']
print(agregar_item_correcto("B")) # ['B']

```

## 3. Empaquetado y Desempaquetado Avanzado: `*args` y `**kwargs`

### 🔹 Concepto Central

* **`*args`** (Arguments): Empaqueta argumentos posicionales adicionales en una **tupla**.
* **`**kwargs`** (Keyword Arguments): Empaqueta argumentos por nombre adicionales en un **diccionario**.

```python
def ejemplo_empaquetado(*args, **kwargs):
    print("args (Tupla):", args)
    print("kwargs (Diccionario):", kwargs)

ejemplo_empaquetado(1, 2, 3, nombre="Alice", activo=True)
# Salida:
# args (Tupla): (1, 2, 3)
# kwargs (Diccionario): {'nombre': 'Alice', 'activo': True}

```

---

### 🔹 Desempaquetado al Invocarse

Puedes usar `*` y `**` al llamar funciones para desarmar colecciones:

```python
def sumar_tres_numeros(a, b, c):
    return a + b + c

mis_numeros = [10, 20, 30]
print(sumar_tres_numeros(*mis_numeros))  # Desempaqueta la lista en a=10, b=20, c=30

mis_datos = {"a": 5, "b": 15, "c": 25}
print(sumar_tres_numeros(**mis_datos))  # Desempaqueta el diccionario

```

---

### 🔹 Orden Estricto de Parámetros en una Firma de Función

Si combinas diferentes tipos de parámetros, debes respetar el siguiente orden:

$$\text{1. Posicionales} \longrightarrow \text{2. *args} \longrightarrow \text{3. Valores por Defecto} \longrightarrow \text{4. **kwargs}$$

```python
def firma_completa(a, b, *args, opcion="default", **kwargs):
    print(f"a: {a}, b: {b}")
    print(f"args: {args}")
    print(f"opcion: {opcion}")
    print(f"kwargs: {kwargs}")

firma_completa(1, 2, 3, 4, 5, opcion="personalizado", x=100, y=200)
# Salida:
# a: 1, b: 2
# args: (3, 4, 5)
# opcion: personalizado
# kwargs: {'x': 100, 'y': 200}

```

---

### 🔹 Caso de Uso Real: Wrappers o Reenvío de Argumentos

`*args` y `**kwargs` permiten construir funciones intermedias que aceptan y reenvían cualquier combinación de parámetros:

```python
def auditoria_wrapper(funcion_objetivo, *args, **kwargs):
    print(f"[LOG] Ejecutando {funcion_objetivo.__name__}...")
    resultado = funcion_objetivo(*args, **kwargs)
    print(f"[LOG] Finalizado con resultado: {resultado}")
    return resultado

def multiplicar(a, b):
    return a * b

auditoria_wrapper(multiplicar, 4, 5)

```

## 4. Ámbito de Variables (Scope) y Regla LEGB

### 🔹 La Regla LEGB

Cuando Python resuelve el valor de una variable dentro de una función, busca en orden estricto dentro de cuatro niveles jerárquicos:

```
[ L ] Local       --> Variables definidas dentro de la función actual.
[ E ] Enclosing   --> Variables en el ámbito de funciones anidadas (padre).
[ G ] Global      --> Variables definidas en el nivel superior del módulo/archivo.
[ B ] Built-in    --> Nombres predefinidos de Python (print, len, int, Exception...).

```

```python
x = "GLOBAL"

def funcion_padre():
    x = "ENCLOSING"
    
    def funcion_hija():
        x = "LOCAL"
        print("Valor de x:", x)  # Busca primero en Local -> Imprime "LOCAL"
    
    funcion_hija()

funcion_padre()

```

---

### 🔹 Modificadores `global` y `nonlocal`

* **`global`**: Permite modificar una variable del nivel global dentro de un ámbito local.
* **`nonlocal`**: Permite modificar una variable de la función contenedora (*enclosing*) dentro de una función anidada.

```python
contador_global = 0

def incrementar_global():
    global contador_global
    contador_global += 1

def funcion_externa():
    conteo = 0
    def funcion_interna():
        nonlocal conteo
        conteo += 1
        return conteo
    return funcion_interna

```

> ⚠️ **Buenas prácticas:** Evita el uso de `global` siempre que sea posible. Las variables globales hacen que el código sea difícil de depurar, probar y mantener. Prefiere pasar datos mediante argumentos y retornar valores.

## 5. Funciones Lambda (Anónimas) y Programación Funcional

### 🔹 Sintaxis y Limitaciones

Una función `lambda` es una función anónima pequeña que se escribe en una sola línea.

$$\text{Sintaxis: } \texttt{lambda } \text{parámetros} : \text{expresión\_única}$$

```python
# Función tradicional
def cuadrado_def(x):
    return x ** 2

# Función Lambda equivalente
cuadrado_lambda = lambda x: x ** 2

print(cuadrado_def(5))     # 25
print(cuadrado_lambda(5))  # 25

```

**Restricciones de `lambda`:**

* Solo puede contener **una única expresión**, cuyo resultado se retorna automáticamente.
* NO puede contener instrucciones o sentencias de bloque (`if/else` complejos de varias líneas, `for`, `while`, `assert`, `pass`, `raise`).

---

### 🔹 Uso Real con `sorted()`, `min()`, `max()`, `map()`, `filter()`

Las lambdas son especialmente útiles para pasar comportamientos como argumento (`key`) en funciones integradas:

```python
# 1. Ordenar una lista de diccionarios por una clave específica
estudiantes = [
    {"nombre": "Ana", "nota": 88},
    {"nombre": "Bob", "nota": 95},
    {"nombre": "Charlie", "nota": 70}
]

# Ordenar por nota de menor a mayor usando el parámetro 'key'
estudiantes_ordenados = sorted(estudiantes, key=lambda est: est["nota"])
print(estudiantes_ordenados)

# 2. Filtrar números pares con filter()
numeros = [1, 2, 3, 4, 5, 6, 7, 8]
pares = list(filter(lambda x: x % 2 == 0, numeros))  # [2, 4, 6, 8]

# 3. Transformar elementos con map()
cuadrados = list(map(lambda x: x ** 2, numeros))    # [1, 4, 9, 16, 25, 36, 49, 64]

```

---

### 🔹 ¿Cuándo usar `lambda` vs. `def`? (PEP 8)

| Criterio | Usar `lambda` | Usar `def` |
| --- | --- | --- |
| **Uso** | Operaciones simples de una línea pasadas como argumento temporal (`key=...`). | Lógica reusable o que abarca múltiples líneas. |
| **Nombramiento** | Generalmente sin nombre (anónima). | Nombre descriptivo en `snake_case`. |
| **PEP 8** | ❌ **Evitar:** `suma = lambda x, y: x + y` |  |



(Asignar una lambda a un nombre invalida su propósito de ser anónima). | ✅ **Preferir:** `def suma(x, y): return x + y` |

## 6. Funciones como Objetos de Primera Clase

En Python, las funciones son **objetos de primera clase** (*First-Class Objects*). Esto significa que se comportan como cualquier otro valor (un entero, una cadena o una lista).

---

### 🔹 1. Asignar funciones a variables

```python
def saludar(nombre):
    return f"Hola, {nombre}"

# Asignamos la función (sin paréntesis) a una nueva variable:
mi_funcion = saludar

# Ahora 'mi_funcion' ejecuta el mismo código:
print(mi_funcion("Carlos"))  # Salida: Hola, Carlos

```

---

### 🔹 2. Pasar funciones como argumentos a otras funciones

Las funciones que aceptan otras funciones como parámetros se denominan **Funciones de Orden Superior** (*Higher-Order Functions*):

```python
def aplicar_operacion(funcion, a, b):
    """Acepta una función como primer argumento y la aplica a 'a' y 'b'."""
    return funcion(a, b)

def multiplicar(x, y):
    return x * y

def restar(x, y):
    return x - y

print(aplicar_operacion(multiplicar, 6, 7)) # Salida: 42
print(aplicar_operacion(restar, 10, 3))      # Salida: 7

```

---

### 🔹 3. Retornar funciones desde dentro de otras funciones

```python
def obtener_multiplicador(factor):
    """Crea y retorna una nueva función configurada con el factor especificado."""
    def multiplicar(numero):
        return numero * factor
    return multiplicar  # Retorna la función interna sin ejecutar

duplicar = obtener_multiplicador(2)
triplicar = obtener_multiplicador(3)

print(duplicar(5))   # Salida: 10
print(triplicar(5))  # Salida: 15

```

## 7. Clausuras (Closures) y Decoradores

### 🔹 ¿Qué es una Clausura (*Closure*)?

Una **Clausura** ocurre cuando una función anidada recuerda y accede a las variables del ámbito de su función contenedora (*enclosing scope*), incluso después de que la función externa ha terminado de ejecutarse.

```python
def crear_contador():
    conteo = 0  # Estado privado retenido
    
    def incrementar():
        nonlocal conteo
        conteo += 1
        return conteo
    
    return incrementar

c1 = crear_contador()
print(c1())  # 1
print(c1())  # 2

c2 = crear_contador()  # Crea un nuevo ámbito con su propio 'conteo'
print(c2())  # 1 (Independiente de c1)

```

---

### 🔹 Decoradores (`@`): Concepto y Sintaxis

Un **Decorador** es una función que toma otra función como argumento, extiende o modifica su comportamiento sin alterar su código fuente, y retorna una nueva función.

```
            [ Función Decoradora ]
                      │
           (Toma la función original)
                      │
                      ▼
┌──────────────────────────────────────────┐
│          Función Wrapper                 │
│  - Código previo (Log, tiempo...)        │
│  - EJECUTA LA FUNCIÓN ORIGINAL           │
│  - Código posterior                      │
└──────────────────────────────────────────┘
                      │
          (Retorna la función envuelta)

```

```python
def mi_decorador(funcion_original):
    def funcion_envoltorio(*args, **kwargs):
        print(">>> Antes de ejecutar la función...")
        resultado = funcion_original(*args, **kwargs)
        print(">>> Después de ejecutar la función...")
        return resultado
    return funcion_envoltorio

# Aplicación con la sintaxis de azúcar 'sugar' @
@mi_decorador
def saludar(nombre):
    print(f"Hola, {nombre}!")

saludar("Muzait")
# Salida:
# >>> Antes de ejecutar la función...
# Hola, Muzait!
# >>> Después de ejecutar la función...

```

---

### 🔹 La Importancia de `@functools.wraps`

Al envolver una función, esta pierde su identidad (su nombre `__name__` y su docstring `__doc__` pasan a ser los de la función envoltorio). Para preservar los metadatos originales, se debe usar `@wraps`:

```python
from functools import wraps

def decorador_profesional(func):
    @wraps(func)  # Preserva __name__ y __doc__ de func
    def wrapper(*args, **kwargs):
        return func(*args, **kwargs)
    return wrapper

@decorador_profesional
def suma_documentada(a: int, b: int) -> int:
    """Suma dos enteros."""
    return a + b

print(suma_documentada.__name__)  # Salida: suma_documentada (Sin @wraps diría 'wrapper')
print(suma_documentada.__doc__)   # Salida: Suma dos enteros.

```

---

### 🔹 Decoradores con Argumentos

Si necesitas pasarle parámetros al propio decorador, requiere un nivel adicional de anidación:

```python
from functools import wraps

def repetir(num_veces):
    def decorador(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            for _ in range(num_veces):
                resultado = func(*args, **kwargs)
            return resultado
        return wrapper
    return decorador

@repetir(num_veces=3)
def avisar():
    print("¡Alerta del sistema!")

avisar()  # Imprime la alerta 3 veces

```

---

### 🔹 Ejemplo Real: Medidor de Tiempo de Ejecución

```python
import time
from functools import wraps

def medir_tiempo(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        inicio = time.perf_counter()
        resultado = func(*args, **kwargs)
        fin = time.perf_counter()
        tiempo_total = fin - inicio
        print(f"⏱️ [{func.__name__}] Tiempo de ejecución: {tiempo_total:.6f} segundos.")
        return resultado
    return wrapper

@medir_tiempo
def simular_proceso_pesado():
    time.sleep(0.5)
    return "Proceso completado"

simular_proceso_pesado()

```

## 8. Generadores y la palabra clave `yield`

### 🔹 `return` vs. `yield` (Evaluación Perezosa / *Lazy Evaluation*)

* **`return`**: Procesa **todos los datos de una vez**, los guarda en memoria y finaliza la función.
* **`yield`**: Pausa la ejecución de la función, devuelve un valor puntual y **conserva su estado interno** para reanudar la ejecución justo en esa línea cuando se le solicite el siguiente valor.

| Propiedad | Función Tradicional (`return`) | Función Generadora (`yield`) |
| --- | --- | --- |
| **Uso de Memoria** | Alto ($O(N)$). Crea la colección completa en RAM. | Mínimo ($O(1)$). Genera un elemento a la vez bajo demanda. |
| **Escalabilidad** | Limitada por la capacidad de la memoria. | Excelente para secuencias infinitas o archivos gigantes. |
| **Retorno** | Un único valor o estructura completa. | Un objeto **Iterador** (Lazy Iterable). |

---

### 🔹 Crear un Generador Personalizado

```python
def contador_generador(maximo):
    n = 1
    while n <= maximo:
        yield n  # Produce el valor y pausa la ejecución
        n += 1

# Al invocar la función, NO se ejecuta el código interno; se retorna un objeto generador:
gen = contador_generador(3)

print(next(gen))  # 1 (Ejecuta hasta el primer yield)
print(next(gen))  # 2 (Reanuda y ejecuta hasta el segundo yield)
print(next(gen))  # 3 (Reanuda y ejecuta hasta el tercer yield)
# print(next(gen))  # ❌ Lanza la excepción StopIteration al agotarse

```

Uso habitual en bucles `for`:

```python
for numero in contador_generador(5):
    print(numero)  # Recorre eficientemente cada valor sin cargarlos todos a la vez

```

---

### 🔹 Expresiones Generadoras vs. List Comprehensions

La sintaxis es idéntica, con la diferencia del tipo de delimitadores utilizados: corchetes `[]` para listas vs. paréntesis `()` para generadores.

```python
import sys

# List Comprehension: Construye los 1,000,000 de enteros inmediatamente en RAM
lista = [x for x in range(1_000_000)]

# Expresión Generadora: Genera una regla para producirlos bajo demanda
generador = (x for x in range(1_000_000))

print(f"Memoria de la Lista:     {sys.getsizeof(lista):,} bytes")       # ~8,000,000 bytes
print(f"Memoria del Generador: {sys.getsizeof(generador):,} bytes")  # ~200 bytes

```

## 9. Anotaciones de Tipo (Type Hints) y Recursividad

### 🔹 Type Hints (PEP 484)

Las anotaciones de tipo mejoran el autocompletado del IDE y permiten realizar análisis estáticos de código (usando herramientas como `mypy`), aunque no imponen restricciones de tipo estrictas en tiempo de ejecución de Python.

```python
from typing import Optional, Union, Callable

def procesar_pago(
    monto: float, 
    moneda: str = "USD", 
    callback: Optional[Callable[[bool], None]] = None
) -> Union[str, bool]:
    """Procesa un pago y ejecuta opcionalmente una función callback."""
    if monto <= 0:
        return False
    
    éxito = True
    if callback:
        callback(éxito)
        
    return f"Pago de {monto} {moneda} procesado exitosamente."

```

---

### 🔹 Recursividad: Estructura Fundamental

Una función es **recursiva** cuando se llama a sí misma dentro de su cuerpo. Toda función recursiva debe contar con dos componentes indispensables:

1. **Caso Base:** La condición de parada que detiene la recursión.
2. **Caso Recursivo:** La llamada a sí misma reduciendo el problema a un subproblema más pequeño.

```python
def factorial(n: int) -> int:
    # 1. Caso Base
    if n <= 1:
        return 1
    # 2. Caso Recursivo
    return n * factorial(n - 1)

print(factorial(5))  # 5 * 4 * 3 * 2 * 1 = 120

```

---

### ⚠️ Trampa Mortal: Exceder la Pila de Llamadas (`RecursionError`)

Si la función recursiva no alcanza el caso base o la profundidad es demasiado grande, se desborda la pila (*Call Stack*):

```python
def funcion_infinita(n):
    return funcion_infinita(n + 1) # ❌ Sin caso base

# funcion_infinita(1)  # RecursionError: maximum recursion depth exceeded in comparison

```

---

### 🔹 Optimización con `@lru_cache`

La recursión naive sin memorización puede generar complejidad exponencial $O(2^N)$ al recalcular submódulos repetidos (como en la secuencia de Fibonacci). Se puede optimizar a $O(N)$ usando `@functools.lru_cache`:

```python
from functools import lru_cache

# Sin caché: fibonacci(35) tomaría varios segundos.
# Con lru_cache: guarda en caché los resultados de las llamadas anteriores.
@lru_cache(maxsize=None)
def fibonacci(n: int) -> int:
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(100))  # Calcula el valor casi instantáneamente

```

## 10. 💡 Patrones y Algoritmos Clásicos con Funciones

### 🔹 Patrón 1: Función Acumuladora / Transformadora

Procesa una secuencia acumulando resultados intermedios.

```python
def calcular_promedio_ponderado(valores: list[float], pesos: list[float]) -> float:
    """Calcula la media ponderada a partir de dos listas de números."""
    if len(valores) != len(pesos) or not valores:
        raise ValueError("Las listas deben tener la misma longitud y no estar vacías.")
    
    suma_productos = sum(v * p for v, p in zip(valores, pesos))
    suma_pesos = sum(pesos)
    
    return suma_productos / suma_pesos

print(calcular_promedio_ponderado([85, 90, 75], [0.3, 0.5, 0.2]))

```

### 🔹 Patrón 2: Validación y Early Return (Retorno Anticipado)

Evita la anidación profunda de bloques `if/else` (Código "Espagueti" o "Anidación en Flecha") validando y retornando errores primero.

```python
# ❌ MAL: Anidación profunda (Difícil de leer)
def registrar_usuario_malo(usuario):
    if usuario is not None:
        if "email" in usuario:
            if "@" in usuario["email"]:
                return "Usuario registrado"
            else:
                return "Email inválido"
        else:
            return "Falta el email"
    else:
        return "Usuario no proporcionado"

# ✅ BIEN: Early Returns (Limpio y legible)
def registrar_usuario_bueno(usuario: dict) -> str:
    if not usuario:
        return "Error: Usuario no proporcionado"
    if "email" not in usuario:
        return "Error: Falta el email"
    if "@" not in usuario["email"]:
        return "Error: Email inválido"
    
    # Flujo principal sin anidación
    return "Usuario registrado exitosamente"

```

### 🔹 Patrón 3: Composición de Funciones

Consiste en pasar la salida de una función directamente como la entrada de otra, encadenando trasformaciones.

```python
def limpiar_texto(texto: str) -> str:
    return texto.strip().lower()

def eliminar_puntuacion(texto: str) -> str:
    import string
    return texto.translate(str.maketrans("", "", string.punctuation))

def tokenizar(texto: str) -> list[str]:
    return texto.split()

# Cadena de composición
def procesar_pipeline_nlp(raw_text: str) -> list[str]:
    paso1 = limpiar_texto(raw_text)
    paso2 = eliminar_puntuacion(paso1)
    tokens = tokenizar(paso2)
    return tokens

print(procesar_pipeline_nlp("   ¡Hola, Mundo! Este es un TEST.   "))
# Salida: ['hola', 'mundo', 'este', 'es', 'un', 'test']

```

### 🔹 Patrón 4: Memoización Manual vs. Automática

La memoización guarda en caché los resultados de funciones costosas basados en sus argumentos de entrada.

```python
# Forma 1: Memoización Manual usando un Diccionario
cache_manual = {}

def factorial_memoizado(n: int) -> int:
    if n in cache_manual:
        return cache_manual[n]
    if n <= 1:
        return 1
    
    resultado = n * factorial_memoizado(n - 1)
    cache_manual[n] = resultado
    return resultado

# Forma 2: Memoización Automática Profesional
from functools import lru_cache

@lru_cache(maxsize=128)
def operacion_costosa(x: int) -> int:
    # Simulación de un cálculo pesado
    return x * x + 2 * x + 1

```

### 🔹 Patrón 5: Currificación o Ejecución Parcial (`functools.partial`)

Permite fijar un número determinado de argumentos de una función para crear una versión más específica de la misma.

```python
from functools import partial

def calcular_impuesto(porcentaje: float, precio: float) -> float:
    return precio * (1 + porcentaje)

# Creamos funciones especializadas fijando el argumento 'porcentaje'
aplicar_iva_mx = partial(calcular_impuesto, 0.16)  # IVA México (16%)
aplicar_iva_es = partial(calcular_impuesto, 0.21)  # IVA España (21%)

precio_base = 100.0

print(f"Precio en México: ${aplicar_iva_mx(precio_base):.2f}") # $116.00
print(f"Precio en España: ${aplicar_iva_es(precio_base):.2f}") # $121.00

```
